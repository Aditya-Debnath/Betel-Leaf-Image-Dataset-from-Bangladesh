```python
# 1. IMPORT LIBRARIES
# ==============================================================================
# Install torchinfo if it's not already available
!pip install torchinfo --quiet

import time
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_curve, auc
from sklearn.preprocessing import label_binarize
import numpy as np
import cv2
import torch
from torch import nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
import torchvision
from torchvision import transforms
from torchinfo import summary
import torchvision.models as models
import torch.optim as optim
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import random
from PIL import Image
from itertools import cycle

# ==============================================================================
# 2. CONFIGURATION
# ==============================================================================
print("--- System Information ---")
try:
    gpu_info = !nvidia-smi --query-gpu=gpu_name --format=csv
    gpu_name = gpu_info[1]
    print(f"GPU: {gpu_name}")
    device = torch.device('cuda')
except Exception as e:
    gpu_name = "No GPU available"
    print(f"GPU: {gpu_name}")
    device = torch.device('cpu')

# --- Variables ---
seed = 42
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

## IMPORTANT: Update this path to your dataset location in Kaggle ##
data_dir = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"

output_dir = "/kaggle/working/processed-dataset"
checkpoints_path = "/kaggle/working/"
os.makedirs(output_dir, exist_ok=True)

# --- Hyperparameters ---
train_batch = 32
test_batch = 16
total_class = 4
learning_rate = 0.0001
decay = 1e-4
initial_epochs = 20 # Epochs for initial training with frozen layers
finetune_epochs = 15 # Epochs for fine-tuning with unfrozen layers
patience = 5
image_size = (299, 299) # InceptionV3 requires 299x299 input

# ==============================================================================
# 3. EXPLORE AND SPLIT THE DATASET
# ==============================================================================
# (This section is identical to your provided code and is included for completeness)
def load_split(split_dir):
    file_paths, labels = [], []
    for class_name in os.listdir(split_dir):
        class_dir = os.path.join(split_dir, class_name)
        if os.path.isdir(class_dir):
            for image_name in os.listdir(class_dir):
                file_paths.append(os.path.join(class_dir, image_name))
                labels.append(class_name)
    return pd.DataFrame({"file_path": file_paths, "label": labels})
data = load_split(data_dir)
train_dataframe, temp_dataframe = train_test_split(data, test_size=0.30, stratify=data['label'], random_state=seed)
validation_dataframe, test_dataframe = train_test_split(temp_dataframe, test_size=0.50, stratify=temp_dataframe['label'], random_state=seed)
# (Visualizations are omitted here for brevity, but they will work with this setup)

# ==============================================================================
# 4. DATA PREPROCESSING (Resize and Save)
# ==============================================================================
def data_preprocess(df, split_name, size, quality=95):
    split_dir = os.path.join(output_dir, split_name)
    os.makedirs(split_dir, exist_ok=True)
    processed_file_paths = []
    for idx, row in tqdm(df.iterrows(), total=len(df), desc=f"Processing {split_name} set"):
        label = row['label']
        label_dir = os.path.join(split_dir, label)
        os.makedirs(label_dir, exist_ok=True)
        img = Image.open(row['file_path']).convert("RGB").resize(size, Image.Resampling.LANCZOS)
        file_name = os.path.basename(row['file_path'])
        save_path = os.path.join(label_dir, file_name)
        img.save(save_path, optimize=True, quality=quality)
        processed_file_paths.append(save_path)
    df['file_path'] = processed_file_paths
    return df
train_dataframe = data_preprocess(train_dataframe, "train", size=image_size)
validation_dataframe = data_preprocess(validation_dataframe, "validation", size=image_size)
test_dataframe = data_preprocess(test_dataframe, "test", size=image_size)

# ==============================================================================
# 5. DATASET CLASS, TRANSFORMS, AND DATALOADERS
# ==============================================================================
class ImageDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe
        self.transform = transform
        unique_labels = sorted(self.dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(unique_labels)}
    def __len__(self): return len(self.dataframe)
    def __getitem__(self, index):
        img_path = self.dataframe.iloc[index]['file_path']
        image = Image.open(img_path).convert("RGB")
        label = self.label_map[self.dataframe.iloc[index]['label']]
        if self.transform: image = self.transform(image)
        return image, label

def get_dataloaders(size):
    normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    train_transform = transforms.Compose([
        transforms.RandomRotation(degrees=15), transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomVerticalFlip(p=0.5), transforms.RandomResizedCrop(size, scale=(0.9, 1.0)),
        transforms.ColorJitter(brightness=0.2, contrast=0.2), transforms.ToTensor(), normalize])
    val_test_transform = transforms.Compose([transforms.Resize(size), transforms.ToTensor(), normalize])
    train_dataset = ImageDataset(train_dataframe, transform=train_transform)
    valid_dataset = ImageDataset(validation_dataframe, transform=val_test_transform)
    test_dataset = ImageDataset(test_dataframe, transform=val_test_transform)
    print("\n--- Label Encoding ---")
    for label, idx in train_dataset.label_map.items(): print(f"{label} -> {idx}")
    train_dataloader = DataLoader(train_dataset, batch_size=train_batch, shuffle=True, num_workers=2)
    validation_dataloader = DataLoader(valid_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    test_dataloader = DataLoader(test_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    return train_dataloader, validation_dataloader, test_dataloader, train_dataset.label_map

train_dataloader, validation_dataloader, test_dataloader, label_map = get_dataloaders(size=image_size)
class_names = list(label_map.keys())

# ==============================================================================
# 6. DEEP LEARNING MODEL (InceptionV3)
# ==============================================================================
print("\n--- Initializing InceptionV3 Model ---")
model = models.inception_v3(weights='IMAGENET1K_V1')
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, total_class)
num_ftrs_aux = model.AuxLogits.fc.in_features
model.AuxLogits.fc = nn.Linear(num_ftrs_aux, total_class)
model = model.to(device)
optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=decay)
criterion = nn.CrossEntropyLoss()
summary(model, input_size=(train_batch, 3, 299, 299))

# ==============================================================================
# 7. MODEL TRAINING AND FINE-TUNING
# ==============================================================================
def train_finetune_model(model, criterion, optimizer, train_loader, val_loader, initial_epochs, finetune_epochs, patience, ckpt_path):
    train_loss_history, val_loss_history = [], []
    train_acc_history, val_acc_history = [], []
    best_val_acc = 0.0
    no_improvement_count = 0
    total_start_time = time.time()
    
    # --- STAGE 1: Initial Training (Feature Extraction) ---
    print("\n--- Stage 1: Training the Classifier (Frozen Base) ---")
    # Freeze all parameters first
    for param in model.parameters():
        param.requires_grad = False
    # Unfreeze the final fully connected layers (main and auxiliary)
    for param in model.fc.parameters():
        param.requires_grad = True
    for param in model.AuxLogits.fc.parameters():
        param.requires_grad = True

    for epoch in range(initial_epochs):
        model.train()
        running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{initial_epochs} [T]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs, aux_outputs = model(inputs)
            loss1 = criterion(outputs, labels)
            loss2 = criterion(aux_outputs, labels)
            loss = loss1 + 0.4 * loss2
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)
        
        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)

        model.eval()
        running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()
        
        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)
        
        print(f"Epoch {epoch+1}/{initial_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 1 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 1 after {patience} epochs with no improvement.")
            break
            
    # --- STAGE 2: Fine-Tuning ---
    print("\n--- Stage 2: Fine-Tuning (Unfrozen Base) ---")
    # Unfreeze all layers for fine-tuning
    for param in model.parameters():
        param.requires_grad = True
    
    # Use a much lower learning rate for fine-tuning
    optimizer_ft = optim.Adam(model.parameters(), lr=learning_rate / 10)
    
    no_improvement_count = 0 # Reset counter

    for epoch in range(finetune_epochs):
        model.train()
        running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{finetune_epochs} [FT]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer_ft.zero_grad()
            outputs, aux_outputs = model(inputs)
            loss1 = criterion(outputs, labels)
            loss2 = criterion(aux_outputs, labels)
            loss = loss1 + 0.4 * loss2
            loss.backward()
            optimizer_ft.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)

        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)
        
        model.eval()
        running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()

        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)
        print(f"Epoch {epoch+1}/{finetune_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 2 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
            print(f"   (No improvement for {no_improvement_count} epochs)")
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 2 after {patience} epochs with no improvement.")
            break
            
    total_end_time = time.time()
    print(f"\n--- Training & Fine-Tuning Finished ---")
    print(f"Total Time: {(total_end_time - total_start_time)/60:.2f} minutes")
    return train_loss_history, train_acc_history, val_loss_history, val_acc_history

# Train and fine-tune the model
train_loss, train_acc, val_loss, val_acc = train_finetune_model(
    model, criterion, optimizer, train_dataloader, validation_dataloader, 
    initial_epochs, finetune_epochs, patience, checkpoints_path
)

# --- Plot Training History ---
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_loss, label='Train Loss'); plt.plot(val_loss, label='Validation Loss')
plt.axvline(x=len(train_loss) - finetune_epochs - 1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs'); plt.ylabel('Loss'); plt.title('Loss over Epochs'); plt.legend()
plt.subplot(1, 2, 2)
plt.plot(train_acc, label='Train Accuracy'); plt.plot(val_acc, label='Validation Accuracy')
plt.axvline(x=len(train_acc) - finetune_epochs - 1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs'); plt.ylabel('Accuracy'); plt.title('Accuracy over Epochs'); plt.legend()
plt.tight_layout(); plt.show()

# ==============================================================================
# 8. EVALUATION
# ==============================================================================
print("\n--- Evaluating Model on Test Set ---")
model.load_state_dict(torch.load(os.path.join(checkpoints_path, "best_model.pt")))
def evaluate_model_full(model, dataloader):
    model.eval(); y_true, y_pred, y_prob = [], [], []
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            probs = F.softmax(outputs, dim=1)
            _, predicted = torch.max(outputs, 1)
            y_true.extend(labels.cpu().numpy()); y_pred.extend(predicted.cpu().numpy()); y_prob.extend(probs.cpu().numpy())
    return np.array(y_true), np.array(y_pred), np.array(y_prob)
y_true, y_pred, y_prob = evaluate_model_full(model, test_dataloader)
print(f"Overall Test Accuracy: {accuracy_score(y_true, y_pred):.3f}")
print("\nClassification Report:"); print(classification_report(y_true, y_pred, target_names=class_names))
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label'); plt.ylabel('True Label'); plt.title('Confusion Matrix'); plt.show()

# --- ROC Curve and AUC Score ---
y_true_binarized = label_binarize(y_true, classes=range(total_class))
fpr, tpr, roc_auc = {}, {}, {}
for i in range(total_class):
    fpr[i], tpr[i], _ = roc_curve(y_true_binarized[:, i], y_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])
plt.figure(figsize=(10, 8)); colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green'])
for i, color in zip(range(total_class), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2, label=f'ROC curve of class {class_names[i]} (AUC = {roc_auc[i]:0.2f})')
plt.plot([0, 1], [0, 1], 'k--', lw=2); plt.xlim([0.0, 1.0]); plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.title('Multi-class ROC Curve for InceptionV3'); plt.legend(loc="lower right"); plt.show()

# ==============================================================================
# 9. EXPLAINABLE AI (Grad-CAM)
# ==============================================================================
class GradCAM:
    def __init__(self, model, target_layer):
        self.model=model; self.target_layer=target_layer; self.gradients=None; self.activations=None; self.hook_handles=[]
        self._register_hooks()
    def _register_hooks(self):
        def f_hook(module, input, output): self.activations = output.detach()
        def b_hook(module, grad_input, grad_output): self.gradients = grad_output[0].detach()
        self.hook_handles.extend([self.target_layer.register_forward_hook(f_hook), self.target_layer.register_backward_hook(b_hook)])
    def remove_hooks(self):
        for handle in self.hook_handles: handle.remove()
    def generate_cam(self, input_tensor, target_class=None):
        self.model.eval(); input_tensor.requires_grad_(True); output = self.model(input_tensor)
        if target_class is None: target_class = output.argmax(dim=1).item()
        self.model.zero_grad(); class_score = output[:, target_class]; class_score.backward()
        weights = torch.mean(self.gradients, dim=(2, 3), keepdim=True)
        cam = torch.sum(weights * self.activations, dim=1).squeeze().cpu().numpy()
        cam = np.maximum(cam, 0); cam = cv2.resize(cam, (299, 299)); cam = (cam - np.min(cam)) / (np.max(cam) - np.min(cam) + 1e-8)
        return cam
def denormalize_image(tensor, mean, std):
    tensor = tensor.clone()
    for t, m, s in zip(tensor, mean, std): t.mul_(s).add_(m)
    return tensor
def show_gradcam(model, dataloader, device, target_layer, classes, num_samples=5):
    gradcam = GradCAM(model, target_layer); data_iter = iter(dataloader)
    plt.figure(figsize=(8, num_samples * 3))
    for i in range(num_samples):
        inputs, labels = next(data_iter); input_tensor = inputs[0].unsqueeze(0).to(device); label = labels[0].item()
        output = model(input_tensor); pred_class = output.argmax(dim=1).item()
        cam = gradcam.generate_cam(input_tensor, pred_class)
        mean_t = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1); std_t = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
        img_denorm = denormalize_image(inputs[0].cpu(), mean_t, std_t); img_np = np.clip(img_denorm.numpy().transpose(1, 2, 0), 0, 1)
        heatmap = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)
        overlay = cv2.addWeighted(cv2.cvtColor(np.uint8(255 * img_np), cv2.COLOR_RGB2BGR), 0.6, heatmap, 0.4, 0)
        plt.subplot(num_samples, 2, 2 * i + 1); plt.imshow(img_np); plt.title(f"Original\nTrue: {classes[label]}"); plt.axis('off')
        plt.subplot(num_samples, 2, 2 * i + 2); plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB)); plt.title(f"Grad-CAM\nPredicted: {classes[pred_class]}"); plt.axis('off')
    plt.tight_layout(); plt.show(); gradcam.remove_hooks()

print("\n--- Grad-CAM Visualization ---")
target_layer = model.Mixed_7c
show_gradcam(model, test_dataloader, device, target_layer, class_names, num_samples=5)
```

```python
# 1. IMPORT LIBRARIES
# ==============================================================================
# Install torchinfo if it's not already available
!pip install torchinfo --quiet

import time
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_curve, auc
from sklearn.preprocessing import label_binarize
import numpy as np
import cv2
import torch
from torch import nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset, random_split
import torchvision
from torchvision import transforms
from torchinfo import summary
import torchvision.models as models
import torch.optim as optim
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import random
from PIL import Image
from itertools import cycle

# ==============================================================================
# 2. CONFIGURATION
# ==============================================================================
print("--- System Information ---")
try:
    gpu_info = !nvidia-smi --query-gpu=gpu_name --format=csv
    gpu_name = gpu_info[1]
    print(f"GPU: {gpu_name}")
    device = torch.device('cuda')
except Exception as e:
    gpu_name = "No GPU available"
    print(f"GPU: {gpu_name}")
    device = torch.device('cpu')

# --- Variables ---
seed = 42
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

## IMPORTANT: Update this path to your dataset location in Kaggle ##
data_dir = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"

output_dir = "/kaggle/working/processed-dataset"
checkpoints_path = "/kaggle/working/"
os.makedirs(output_dir, exist_ok=True)

# --- Hyperparameters ---
train_batch = 32
test_batch = 16
total_class = 4
learning_rate = 0.0001
decay = 1e-4
initial_epochs = 20
finetune_epochs = 15
patience = 5
image_size = (224, 224)

# ==============================================================================
# 3. EXPLORE AND SPLIT THE DATASET
# ==============================================================================
def load_split(split_dir):
    file_paths, labels = [], []
    for class_name in os.listdir(split_dir):
        class_dir = os.path.join(split_dir, class_name)
        if os.path.isdir(class_dir):
            for image_name in os.listdir(class_dir):
                file_paths.append(os.path.join(class_dir, image_name))
                labels.append(class_name)
    return pd.DataFrame({"file_path": file_paths, "label": labels})

data = load_split(data_dir)
train_dataframe, temp_dataframe = train_test_split(data, test_size=0.30, stratify=data['label'], random_state=seed)
validation_dataframe, test_dataframe = train_test_split(temp_dataframe, test_size=0.50, stratify=temp_dataframe['label'], random_state=seed)

# (Visualization code is omitted here for brevity but can be added back if needed)

# ==============================================================================
# 4. DATA PREPROCESSING (Resize and Save)
# ==============================================================================
def data_preprocess(df, split_name, size, quality=95):
    split_dir = os.path.join(output_dir, split_name)
    os.makedirs(split_dir, exist_ok=True)
    processed_file_paths = []
    for idx, row in tqdm(df.iterrows(), total=len(df), desc=f"Processing {split_name} set"):
        label = row['label']
        label_dir = os.path.join(split_dir, label)
        os.makedirs(label_dir, exist_ok=True)
        img = Image.open(row['file_path']).convert("RGB").resize(size, Image.Resampling.LANCZOS)
        file_name = os.path.basename(row['file_path'])
        save_path = os.path.join(label_dir, file_name)
        img.save(save_path, optimize=True, quality=quality)
        processed_file_paths.append(save_path)
    df['file_path'] = processed_file_paths
    return df

train_dataframe = data_preprocess(train_dataframe, "train", size=image_size)
validation_dataframe = data_preprocess(validation_dataframe, "validation", size=image_size)
test_dataframe = data_preprocess(test_dataframe, "test", size=image_size)

# ==============================================================================
# 5. DATASET CLASS, TRANSFORMS, AND DATALOADERS
# ==============================================================================
class ImageDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe
        self.transform = transform
        unique_labels = sorted(self.dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(unique_labels)}
    def __len__(self):
        return len(self.dataframe)
    def __getitem__(self, index):
        img_path = self.dataframe.iloc[index]['file_path']
        image = Image.open(img_path).convert("RGB")
        label = self.label_map[self.dataframe.iloc[index]['label']]
        if self.transform:
            image = self.transform(image)
        return image, label

def get_dataloaders(size):
    normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    train_transform = transforms.Compose([
        transforms.RandomRotation(degrees=15),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomVerticalFlip(p=0.5),
        transforms.RandomResizedCrop(size, scale=(0.9, 1.0)),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.ToTensor(),
        normalize
    ])
    val_test_transform = transforms.Compose([transforms.Resize(size), transforms.ToTensor(), normalize])
    train_dataset = ImageDataset(train_dataframe, transform=train_transform)
    valid_dataset = ImageDataset(validation_dataframe, transform=val_test_transform)
    test_dataset = ImageDataset(test_dataframe, transform=val_test_transform)
    print("\n--- Label Encoding ---")
    for label, idx in train_dataset.label_map.items(): print(f"{label} -> {idx}")
    train_dataloader = DataLoader(train_dataset, batch_size=train_batch, shuffle=True, num_workers=2)
    validation_dataloader = DataLoader(valid_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    test_dataloader = DataLoader(test_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    return train_dataloader, validation_dataloader, test_dataloader, train_dataset.label_map

train_dataloader, validation_dataloader, test_dataloader, label_map = get_dataloaders(size=image_size)
class_names = list(label_map.keys())

# ==============================================================================
# 6. DEEP LEARNING MODEL (ResNet50)
# ==============================================================================
print("\n--- Initializing ResNet50 Model ---")
model = models.resnet50(weights='IMAGENET1K_V1')
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, total_class)
model = model.to(device)
optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=decay)
criterion = nn.CrossEntropyLoss()
summary(model, input_size=(train_batch, 3, 224, 224))

# ==============================================================================
# 7. MODEL TRAINING AND FINE-TUNING
# ==============================================================================
def train_finetune_model(model, criterion, optimizer, train_loader, val_loader, initial_epochs, finetune_epochs, patience, ckpt_path):
    train_loss_history, val_loss_history = [], []
    train_acc_history, val_acc_history = [], []
    best_val_acc = 0.0
    no_improvement_count = 0
    total_start_time = time.time()
    
    # --- STAGE 1: Initial Training (Feature Extraction) ---
    print("\n--- Stage 1: Training the Classifier (Frozen Base) ---")
    for param in model.parameters(): param.requires_grad = False
    for param in model.fc.parameters(): param.requires_grad = True
    for epoch in range(initial_epochs):
        model.train(); running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{initial_epochs} [T]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward(); optimizer.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)
        
        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)
        
        model.eval(); running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()
        
        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)
        print(f"Epoch {epoch+1}/{initial_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 1 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else: no_improvement_count += 1
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 1 after {patience} epochs with no improvement.")
            break
            
    # --- STAGE 2: Fine-Tuning ---
    print("\n--- Stage 2: Fine-Tuning (Unfrozen Base) ---")
    for param in model.parameters(): param.requires_grad = True
    optimizer_ft = optim.Adam(model.parameters(), lr=learning_rate / 10)
    no_improvement_count = 0
    for epoch in range(finetune_epochs):
        model.train(); running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{finetune_epochs} [FT]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer_ft.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward(); optimizer_ft.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)
        
        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)
        
        model.eval(); running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()
        
        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)
        print(f"Epoch {epoch+1}/{finetune_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 2 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
            print(f"   (No improvement for {no_improvement_count} epochs)")
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 2 after {patience} epochs with no improvement.")
            break
            
    total_end_time = time.time()
    print(f"\n--- Training & Fine-Tuning Finished ---")
    print(f"Total Time: {(total_end_time - total_start_time)/60:.2f} minutes")
    return train_loss_history, train_acc_history, val_loss_history, val_acc_history

# Train and fine-tune the model
train_loss, train_acc, val_loss, val_acc = train_finetune_model(
    model, criterion, optimizer, train_dataloader, validation_dataloader, 
    initial_epochs, finetune_epochs, patience, checkpoints_path
)

# --- Plot Training History ---
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_loss, label='Train Loss'); plt.plot(val_loss, label='Validation Loss')
plt.axvline(x=len(train_loss) - finetune_epochs -1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs'); plt.ylabel('Loss'); plt.title('Loss over Epochs'); plt.legend()
plt.subplot(1, 2, 2)
plt.plot(train_acc, label='Train Accuracy'); plt.plot(val_acc, label='Validation Accuracy')
plt.axvline(x=len(train_acc) - finetune_epochs - 1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs'); plt.ylabel('Accuracy'); plt.title('Accuracy over Epochs'); plt.legend()
plt.tight_layout(); plt.show()

# ==============================================================================
# 8. EVALUATION
# ==============================================================================
print("\n--- Evaluating Model on Test Set ---")
model.load_state_dict(torch.load(os.path.join(checkpoints_path, "best_model.pt")))
def evaluate_model_full(model, dataloader):
    model.eval(); y_true, y_pred, y_prob = [], [], []
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            probs = F.softmax(outputs, dim=1)
            _, predicted = torch.max(outputs, 1)
            y_true.extend(labels.cpu().numpy()); y_pred.extend(predicted.cpu().numpy()); y_prob.extend(probs.cpu().numpy())
    return np.array(y_true), np.array(y_pred), np.array(y_prob)
y_true, y_pred, y_prob = evaluate_model_full(model, test_dataloader)
print(f"Overall Test Accuracy: {accuracy_score(y_true, y_pred):.3f}")
print("\nClassification Report:"); print(classification_report(y_true, y_pred, target_names=class_names))
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label'); plt.ylabel('True Label'); plt.title('Confusion Matrix'); plt.show()

# --- ROC Curve and AUC Score ---
y_true_binarized = label_binarize(y_true, classes=range(total_class))
fpr, tpr, roc_auc = {}, {}, {}
for i in range(total_class):
    fpr[i], tpr[i], _ = roc_curve(y_true_binarized[:, i], y_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])
fpr["micro"], tpr["micro"], _ = roc_curve(y_true_binarized.ravel(), y_prob.ravel())
roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])
plt.figure(figsize=(10, 8)); colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green'])
for i, color in zip(range(total_class), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2, label=f'ROC curve of class {class_names[i]} (area = {roc_auc[i]:0.2f})')
plt.plot([0, 1], [0, 1], 'k--', lw=2); plt.xlim([0.0, 1.0]); plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.title('Multi-class ROC Curve'); plt.legend(loc="lower right"); plt.show()

# ==============================================================================
# 9. EXPLAINABLE AI (Grad-CAM)
# ==============================================================================
class GradCAM:
    def __init__(self, model, target_layer):
        self.model=model; self.target_layer=target_layer; self.gradients=None; self.activations=None; self.hook_handles=[]
        self._register_hooks()
    def _register_hooks(self):
        def f_hook(module, input, output): self.activations = output.detach()
        def b_hook(module, grad_input, grad_output): self.gradients = grad_output[0].detach()
        self.hook_handles.extend([self.target_layer.register_forward_hook(f_hook), self.target_layer.register_backward_hook(b_hook)])
    def remove_hooks(self):
        for handle in self.hook_handles: handle.remove()
    def generate_cam(self, input_tensor, target_class=None):
        self.model.eval(); input_tensor.requires_grad_(True); output = self.model(input_tensor)
        if target_class is None: target_class = output.argmax(dim=1).item()
        self.model.zero_grad(); class_score = output[:, target_class]; class_score.backward()
        weights = torch.mean(self.gradients, dim=(2, 3), keepdim=True)
        cam = torch.sum(weights * self.activations, dim=1).squeeze().cpu().numpy()
        cam = np.maximum(cam, 0); cam = cv2.resize(cam, (224, 224)); cam = (cam - np.min(cam)) / (np.max(cam) - np.min(cam) + 1e-8)
        return cam
def denormalize_image(tensor, mean, std):
    tensor = tensor.clone()
    for t, m, s in zip(tensor, mean, std): t.mul_(s).add_(m)
    return tensor
def show_gradcam(model, dataloader, device, target_layer, classes, num_samples=5):
    gradcam = GradCAM(model, target_layer); data_iter = iter(dataloader)
    plt.figure(figsize=(8, num_samples * 3))
    for i in range(num_samples):
        inputs, labels = next(data_iter); input_tensor = inputs[0].unsqueeze(0).to(device); label = labels[0].item()
        output = model(input_tensor); pred_class = output.argmax(dim=1).item()
        cam = gradcam.generate_cam(input_tensor, pred_class)
        mean_t = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1); std_t = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
        img_denorm = denormalize_image(inputs[0].cpu(), mean_t, std_t); img_np = np.clip(img_denorm.numpy().transpose(1, 2, 0), 0, 1)
        heatmap = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)
        overlay = cv2.addWeighted(cv2.cvtColor(np.uint8(255 * img_np), cv2.COLOR_RGB2BGR), 0.6, heatmap, 0.4, 0)
        plt.subplot(num_samples, 2, 2 * i + 1); plt.imshow(img_np); plt.title(f"Original\nTrue: {classes[label]}"); plt.axis('off')
        plt.subplot(num_samples, 2, 2 * i + 2); plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB)); plt.title(f"Grad-CAM\nPredicted: {classes[pred_class]}"); plt.axis('off')
    plt.tight_layout(); plt.show(); gradcam.remove_hooks()

print("\n--- Grad-CAM Visualization ---")
target_layer = model.layer4[-1].conv3 # Last conv layer of the final block in ResNet50
show_gradcam(model, test_dataloader, device, target_layer, class_names, num_samples=5)
```

```python
# 1. IMPORT LIBRARIES
# ==============================================================================
# Install torchinfo if it's not already available
!pip install torchinfo --quiet

import time
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_curve, auc
from sklearn.preprocessing import label_binarize
import numpy as np
import cv2
import torch
from torch import nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
import torchvision
from torchvision import transforms
from torchinfo import summary
import torchvision.models as models
import torch.optim as optim
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import random
from PIL import Image
from itertools import cycle

# ==============================================================================
# 2. CONFIGURATION
# ==============================================================================
print("--- System Information ---")
try:
    gpu_info = !nvidia-smi --query-gpu=gpu_name --format=csv
    gpu_name = gpu_info[1]
    print(f"GPU: {gpu_name}")
    device = torch.device('cuda')
except Exception as e:
    gpu_name = "No GPU available"
    print(f"GPU: {gpu_name}")
    device = torch.device('cpu')

# --- Variables ---
seed = 42
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

## IMPORTANT: Update this path to your dataset location in Kaggle ##
data_dir = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"

output_dir = "/kaggle/working/processed-dataset"
checkpoints_path = "/kaggle/working/"
os.makedirs(output_dir, exist_ok=True)

# --- Hyperparameters ---
train_batch = 32
test_batch = 16
total_class = 4
learning_rate = 0.0001
decay = 1e-4
initial_epochs = 20 # Epochs for initial training with frozen layers
finetune_epochs = 15 # Epochs for fine-tuning
patience = 5
image_size = (224, 224)

# ==============================================================================
# 3. EXPLORE AND SPLIT THE DATASET
# ==============================================================================
def load_split(split_dir):
    file_paths, labels = [], []
    for class_name in os.listdir(split_dir):
        class_dir = os.path.join(split_dir, class_name)
        if os.path.isdir(class_dir):
            for image_name in os.listdir(class_dir):
                file_paths.append(os.path.join(class_dir, image_name))
                labels.append(class_name)
    return pd.DataFrame({"file_path": file_paths, "label": labels})

data = load_split(data_dir)
train_dataframe, temp_dataframe = train_test_split(data, test_size=0.30, stratify=data['label'], random_state=seed)
validation_dataframe, test_dataframe = train_test_split(temp_dataframe, test_size=0.50, stratify=temp_dataframe['label'], random_state=seed)

# --- Visualize Dataset Split Ratios ---
sizes = {
    "Train": len(train_dataframe),
    "Validation": len(validation_dataframe),
    "Test": len(test_dataframe),
}

total = sum(sizes.values())
ratios = {k: (v / total) * 100 for k, v in sizes.items()}

print("\n--- Dataset Split Ratios ---")
for split in sizes.keys():
    print(f"{split}: {sizes[split]} images ({ratios[split]:.2f}%)")

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].pie(
    sizes.values(),
    labels=sizes.keys(),
    autopct="%.1f%%",
    startangle=90,
    colors=["#a6cee3", "#1f78b4", "#b2df8a"]
)
axes[0].set_title("Dataset Split Pie Chart")

axes[1].bar(sizes.keys(), sizes.values(), color=["#a6cee3", "#1f78b4", "#b2df8a"])
axes[1].set_ylabel("Number of Images")
axes[1].set_title("Dataset Split Bar Chart")
y_max = max(sizes.values()) * 1.1
axes[1].set_ylim(0, y_max)
for i, v in enumerate(sizes.values()):
    axes[1].text(i, v + total*0.01, str(v), ha='center')  
plt.tight_layout()
plt.show()

# --- Visualize Class Distribution ---
def class_distribution_plot(train_df, test_df, val_df):
    datasets = { "Train": train_df, "Validation": val_df, "Test": test_df }
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle('Class Distribution Across Datasets', fontsize=16)
    
    for ax, (split_name, df) in zip(axes, datasets.items()):
        class_counts = df['label'].value_counts().sort_index()
        class_counts.plot(kind='bar', ax=ax, color=sns.color_palette("viridis", len(class_counts)))
        ax.set_xlabel('Classes')
        ax.set_ylabel('Number of Images')
        ax.set_title(f'{split_name} Set')
        ax.set_xticklabels(class_counts.index, rotation=45, ha='right')
        
        for i, count in enumerate(class_counts):
            ax.text(i, count + 2, str(count), ha='center')
        ax.set_ylim(0, max(class_counts) * 1.2)
    
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

class_distribution_plot(train_dataframe, test_dataframe, validation_dataframe)

# ==============================================================================
# 4. DATA PREPROCESSING (Resize and Save)
# ==============================================================================
def data_preprocess(df, split_name, size, quality=95):
    split_dir = os.path.join(output_dir, split_name)
    os.makedirs(split_dir, exist_ok=True)
    processed_file_paths = []
    for idx, row in tqdm(df.iterrows(), total=len(df), desc=f"Processing {split_name} set"):
        label = row['label']
        label_dir = os.path.join(split_dir, label)
        os.makedirs(label_dir, exist_ok=True)
        img = Image.open(row['file_path']).convert("RGB").resize(size, Image.Resampling.LANCZOS)
        file_name = os.path.basename(row['file_path'])
        save_path = os.path.join(label_dir, file_name)
        img.save(save_path, optimize=True, quality=quality)
        processed_file_paths.append(save_path)
    df['file_path'] = processed_file_paths
    return df

train_dataframe = data_preprocess(train_dataframe, "train", size=image_size)
validation_dataframe = data_preprocess(validation_dataframe, "validation", size=image_size)
test_dataframe = data_preprocess(test_dataframe, "test", size=image_size)

# ==============================================================================
# 5. DATASET CLASS, TRANSFORMS, AND DATALOADERS
# ==============================================================================
class ImageDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe
        self.transform = transform
        unique_labels = sorted(self.dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(unique_labels)}
    def __len__(self): return len(self.dataframe)
    def __getitem__(self, index):
        img_path = self.dataframe.iloc[index]['file_path']
        image = Image.open(img_path).convert("RGB")
        label = self.label_map[self.dataframe.iloc[index]['label']]
        if self.transform: image = self.transform(image)
        return image, label

def get_dataloaders(size):
    normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    train_transform = transforms.Compose([
        transforms.RandomRotation(degrees=15), transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomVerticalFlip(p=0.5), transforms.RandomResizedCrop(size, scale=(0.9, 1.0)),
        transforms.ColorJitter(brightness=0.2, contrast=0.2), transforms.ToTensor(), normalize])
    val_test_transform = transforms.Compose([transforms.Resize(size), transforms.ToTensor(), normalize])
    train_dataset = ImageDataset(train_dataframe, transform=train_transform)
    valid_dataset = ImageDataset(validation_dataframe, transform=val_test_transform)
    test_dataset = ImageDataset(test_dataframe, transform=val_test_transform)
    print("\n--- Label Encoding ---")
    for label, idx in train_dataset.label_map.items(): print(f"{label} -> {idx}")
    train_dataloader = DataLoader(train_dataset, batch_size=train_batch, shuffle=True, num_workers=2)
    validation_dataloader = DataLoader(valid_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    test_dataloader = DataLoader(test_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    return train_dataloader, validation_dataloader, test_dataloader, train_dataset.label_map

train_dataloader, validation_dataloader, test_dataloader, label_map = get_dataloaders(size=image_size)
class_names = list(label_map.keys())

# ==============================================================================
# 6. DEEP LEARNING MODEL (MobileNetV2)
# ==============================================================================
print("\n--- Initializing MobileNetV2 Model ---")
model = models.mobilenet_v2(weights='IMAGENET1K_V1')
num_ftrs = model.classifier[1].in_features
model.classifier[1] = nn.Linear(num_ftrs, total_class)
model = model.to(device)
optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=decay)
criterion = nn.CrossEntropyLoss()
summary(model, input_size=(train_batch, 3, 224, 224))

# ==============================================================================
# 7. MODEL TRAINING AND FINE-TUNING
# ==============================================================================
def train_finetune_model(model, criterion, optimizer, train_loader, val_loader, initial_epochs, finetune_epochs, patience, ckpt_path):
    train_loss_history, val_loss_history = [], []
    train_acc_history, val_acc_history = [], []
    best_val_acc = 0.0
    no_improvement_count = 0
    total_start_time = time.time()
    
    # --- STAGE 1: Initial Training (Feature Extraction) ---
    print("\n--- Stage 1: Training the Classifier (Frozen Base) ---")
    for param in model.features.parameters(): param.requires_grad = False
    for param in model.classifier.parameters(): param.requires_grad = True
    for epoch in range(initial_epochs):
        model.train(); running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{initial_epochs} [T]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad(); outputs = model(inputs); loss = criterion(outputs, labels); loss.backward(); optimizer.step()
            running_loss += loss.item() * inputs.size(0); _, predicted = torch.max(outputs, 1); total_train += labels.size(0); correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)
        epoch_train_loss = running_loss / total_train; epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss); train_acc_history.append(epoch_train_acc)
        
        model.eval(); running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs); loss = criterion(outputs, labels); running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1); total_val += labels.size(0); correct_val += (predicted == labels).sum().item()
        
        epoch_val_loss = running_val_loss / total_val; epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss); val_acc_history.append(epoch_val_acc)
        print(f"Epoch {epoch+1}/{initial_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 1 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else: no_improvement_count += 1
        if no_improvement_count >= patience: print(f"🛑 Early stopping Stage 1 after {patience} epochs with no improvement."); break
            
    # --- STAGE 2: Fine-Tuning ---
    print("\n--- Stage 2: Fine-Tuning (Unfrozen Base) ---")
    for param in model.parameters(): param.requires_grad = True
    optimizer_ft = optim.Adam(model.parameters(), lr=learning_rate / 10)
    no_improvement_count = 0
    for epoch in range(finetune_epochs):
        model.train(); running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{finetune_epochs} [FT]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer_ft.zero_grad(); outputs = model(inputs); loss = criterion(outputs, labels); loss.backward(); optimizer_ft.step()
            running_loss += loss.item() * inputs.size(0); _, predicted = torch.max(outputs, 1); total_train += labels.size(0); correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)

        epoch_train_loss = running_loss / total_train; epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss); train_acc_history.append(epoch_train_acc)
        
        model.eval(); running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs); loss = criterion(outputs, labels); running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1); total_val += labels.size(0); correct_val += (predicted == labels).sum().item()

        epoch_val_loss = running_val_loss / total_val; epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss); val_acc_history.append(epoch_val_acc)
        print(f"Epoch {epoch+1}/{finetune_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 2 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
            print(f"   (No improvement for {no_improvement_count} epochs)")
        if no_improvement_count >= patience: print(f"🛑 Early stopping Stage 2 after {patience} epochs with no improvement."); break
            
    total_end_time = time.time()
    print(f"\n--- Training & Fine-Tuning Finished ---")
    print(f"Total Time: {(total_end_time - total_start_time)/60:.2f} minutes")
    return train_loss_history, train_acc_history, val_loss_history, val_acc_history

# Train and fine-tune the model
train_loss, train_acc, val_loss, val_acc = train_finetune_model(
    model, criterion, optimizer, train_dataloader, validation_dataloader, 
    initial_epochs, finetune_epochs, patience, checkpoints_path
)

# --- Plot Training History ---
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_loss, label='Train Loss'); plt.plot(val_loss, label='Validation Loss')
plt.axvline(x=len(train_loss) - finetune_epochs - 1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs'); plt.ylabel('Loss'); plt.title('Loss over Epochs'); plt.legend()
plt.subplot(1, 2, 2)
plt.plot(train_acc, label='Train Accuracy'); plt.plot(val_acc, label='Validation Accuracy')
plt.axvline(x=len(train_acc) - finetune_epochs - 1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs'); plt.ylabel('Accuracy'); plt.title('Accuracy over Epochs'); plt.legend()
plt.tight_layout(); plt.show()

# ==============================================================================
# 8. EVALUATION
# ==============================================================================
print("\n--- Evaluating Model on Test Set ---")
model.load_state_dict(torch.load(os.path.join(checkpoints_path, "best_model.pt")))
def evaluate_model_full(model, dataloader):
    model.eval(); y_true, y_pred, y_prob = [], [], []
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            probs = F.softmax(outputs, dim=1)
            _, predicted = torch.max(outputs, 1)
            y_true.extend(labels.cpu().numpy()); y_pred.extend(predicted.cpu().numpy()); y_prob.extend(probs.cpu().numpy())
    return np.array(y_true), np.array(y_pred), np.array(y_prob)

y_true, y_pred, y_prob = evaluate_model_full(model, test_dataloader)
print(f"Overall Test Accuracy: {accuracy_score(y_true, y_pred):.3f}")
print("\nClassification Report:"); print(classification_report(y_true, y_pred, target_names=class_names))
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label'); plt.ylabel('True Label'); plt.title('Confusion Matrix'); plt.show()

# --- ROC Curve and AUC Score ---
y_true_binarized = label_binarize(y_true, classes=range(total_class))
fpr, tpr, roc_auc = {}, {}, {}
for i in range(total_class):
    fpr[i], tpr[i], _ = roc_curve(y_true_binarized[:, i], y_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])
plt.figure(figsize=(10, 8)); colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green'])
for i, color in zip(range(total_class), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2, label=f'ROC curve of class {class_names[i]} (AUC = {roc_auc[i]:0.2f})')
plt.plot([0, 1], [0, 1], 'k--', lw=2); plt.xlim([0.0, 1.0]); plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.title('Multi-class ROC Curve for MobileNetV2'); plt.legend(loc="lower right"); plt.show()

# ==============================================================================
# 9. EXPLAINABLE AI (Grad-CAM)
# ==============================================================================
class GradCAM:
    def __init__(self, model, target_layer):
        self.model=model; self.target_layer=target_layer; self.gradients=None; self.activations=None; self.hook_handles=[]
        self._register_hooks()
    def _register_hooks(self):
        def f_hook(module, input, output): self.activations = output.detach()
        def b_hook(module, grad_input, grad_output): self.gradients = grad_output[0].detach()
        self.hook_handles.extend([self.target_layer.register_forward_hook(f_hook), self.target_layer.register_backward_hook(b_hook)])
    def remove_hooks(self):
        for handle in self.hook_handles: handle.remove()
    def generate_cam(self, input_tensor, target_class=None):
        self.model.eval(); input_tensor.requires_grad_(True); output = self.model(input_tensor)
        if target_class is None: target_class = output.argmax(dim=1).item()
        self.model.zero_grad(); class_score = output[:, target_class]; class_score.backward()
        weights = torch.mean(self.gradients, dim=(2, 3), keepdim=True)
        cam = torch.sum(weights * self.activations, dim=1).squeeze().cpu().numpy()
        cam = np.maximum(cam, 0); cam = cv2.resize(cam, (224, 224)); cam = (cam - np.min(cam)) / (np.max(cam) - np.min(cam) + 1e-8)
        return cam
def denormalize_image(tensor, mean, std):
    tensor = tensor.clone()
    for t, m, s in zip(tensor, mean, std): t.mul_(s).add_(m)
    return tensor
def show_gradcam(model, dataloader, device, target_layer, classes, num_samples=5):
    gradcam = GradCAM(model, target_layer); data_iter = iter(dataloader)
    plt.figure(figsize=(8, num_samples * 3))
    for i in range(num_samples):
        inputs, labels = next(data_iter); input_tensor = inputs[0].unsqueeze(0).to(device); label = labels[0].item()
        output = model(input_tensor); pred_class = output.argmax(dim=1).item()
        cam = gradcam.generate_cam(input_tensor, pred_class)
        mean_t = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1); std_t = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
        img_denorm = denormalize_image(inputs[0].cpu(), mean_t, std_t); img_np = np.clip(img_denorm.numpy().transpose(1, 2, 0), 0, 1)
        heatmap = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)
        overlay = cv2.addWeighted(cv2.cvtColor(np.uint8(255 * img_np), cv2.COLOR_RGB2BGR), 0.6, heatmap, 0.4, 0)
        plt.subplot(num_samples, 2, 2 * i + 1); plt.imshow(img_np); plt.title(f"Original\nTrue: {classes[label]}"); plt.axis('off')
        plt.subplot(num_samples, 2, 2 * i + 2); plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB)); plt.title(f"Grad-CAM\nPredicted: {classes[pred_class]}"); plt.axis('off')
    plt.tight_layout(); plt.show(); gradcam.remove_hooks()

print("\n--- Grad-CAM Visualization ---")
target_layer = model.features[-1] 
show_gradcam(model, test_dataloader, device, target_layer, class_names, num_samples=5)
```

```python
# ==============================================================================
# 1. IMPORT LIBRARIES
# ==============================================================================
# Install torchinfo if it's not already available
!pip install torchinfo --quiet

import time
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import numpy as np
import cv2
import torch
from torch import nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
import torchvision
from torchvision import transforms
from torchinfo import summary
import torchvision.models as models
import torch.optim as optim
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import random
from PIL import Image

# ==============================================================================
# 2. CONFIGURATION
# ==============================================================================
print("--- System Information ---")
try:
    gpu_info = !nvidia-smi --query-gpu=gpu_name --format=csv
    gpu_name = gpu_info[1]
    print(f"GPU: {gpu_name}")
    device = torch.device('cuda')
except Exception as e:
    gpu_name = "No GPU available"
    print(f"GPU: {gpu_name}")
    device = torch.device('cpu')

# --- Variables ---
seed = 42
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

## IMPORTANT: Update this path to your dataset location in Kaggle ##
data_dir = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"

output_dir = "/kaggle/working/processed-dataset"
checkpoints_path = "/kaggle/working/"
os.makedirs(output_dir, exist_ok=True)

# --- Hyperparameters ---
train_batch = 32
test_batch = 16
total_class = 4
learning_rate = 0.0001
decay = 1e-4
initial_epochs = 20 # Epochs for initial training with frozen layers
finetune_epochs = 15 # Epochs for fine-tuning with unfrozen layers
patience = 5
image_size = (224, 224) # AlexNet uses 224x224 input

# ==============================================================================
# 3. EXPLORE AND SPLIT THE DATASET
# ==============================================================================
def load_split(split_dir):
    file_paths, labels = [], []
    for class_name in os.listdir(split_dir):
        class_dir = os.path.join(split_dir, class_name)
        if os.path.isdir(class_dir):
            for image_name in os.listdir(class_dir):
                file_paths.append(os.path.join(class_dir, image_name))
                labels.append(class_name)
    return pd.DataFrame({"file_path": file_paths, "label": labels})

data = load_split(data_dir)
train_dataframe, temp_dataframe = train_test_split(data, test_size=0.30, stratify=data['label'], random_state=seed)
validation_dataframe, test_dataframe = train_test_split(temp_dataframe, test_size=0.50, stratify=temp_dataframe['label'], random_state=seed)

# --- Visualize Dataset Split Ratios ---
sizes = {"Train": len(train_dataframe), "Validation": len(validation_dataframe), "Test": len(test_dataframe)}
total = sum(sizes.values())
ratios = {k: (v / total) * 100 for k, v in sizes.items()}

print("\n--- Dataset Split Ratios ---")
for split in sizes.keys():
    print(f"{split}: {sizes[split]} images ({ratios[split]:.2f}%)")

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].pie(sizes.values(), labels=sizes.keys(), autopct="%.1f%%", startangle=90, colors=["#a6cee3", "#1f78b4", "#b2df8a"])
axes[0].set_title("Dataset Split Pie Chart")
axes[1].bar(sizes.keys(), sizes.values(), color=["#a6cee3", "#1f78b4", "#b2df8a"])
axes[1].set_ylabel("Number of Images")
axes[1].set_title("Dataset Split Bar Chart")
y_max = max(sizes.values()) * 1.1
axes[1].set_ylim(0, y_max)
for i, v in enumerate(sizes.values()):
    axes[1].text(i, v + total*0.01, str(v), ha='center')  
plt.tight_layout()
plt.show()

# --- Visualize Class Distribution ---
def class_distribution_plot(train_df, test_df, val_df):
    datasets = { "Train": train_df, "Validation": val_df, "Test": test_df }
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle('Class Distribution Across Datasets', fontsize=16)
    
    for ax, (split_name, df) in zip(axes, datasets.items()):
        class_counts = df['label'].value_counts().sort_index()
        class_counts.plot(kind='bar', ax=ax, color=sns.color_palette("viridis", len(class_counts)))
        ax.set_xlabel('Classes')
        ax.set_ylabel('Number of Images')
        ax.set_title(f'{split_name} Set')
        ax.set_xticklabels(class_counts.index, rotation=45, ha='right')
        
        for i, count in enumerate(class_counts):
            ax.text(i, count + 2, str(count), ha='center')
        ax.set_ylim(0, max(class_counts) * 1.2)
    
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

class_distribution_plot(train_dataframe, test_dataframe, validation_dataframe)

# ==============================================================================
# 4. DATA PREPROCESSING (Resize and Save)
# ==============================================================================
def data_preprocess(df, split_name, size, quality=95):
    split_dir = os.path.join(output_dir, split_name)
    os.makedirs(split_dir, exist_ok=True)
    processed_file_paths = []
    for idx, row in tqdm(df.iterrows(), total=len(df), desc=f"Processing {split_name} set"):
        label = row['label']
        label_dir = os.path.join(split_dir, label)
        os.makedirs(label_dir, exist_ok=True)
        img = Image.open(row['file_path']).convert("RGB")
        img = img.resize(size, Image.Resampling.LANCZOS)
        file_name = os.path.basename(row['file_path'])
        save_path = os.path.join(label_dir, file_name)
        img.save(save_path, optimize=True, quality=quality)
        processed_file_paths.append(save_path)
    df['file_path'] = processed_file_paths
    return df

train_dataframe = data_preprocess(train_dataframe, "train", size=image_size)
validation_dataframe = data_preprocess(validation_dataframe, "validation", size=image_size)
test_dataframe = data_preprocess(test_dataframe, "test", size=image_size)

# ==============================================================================
# 5. DATASET CLASS, TRANSFORMS, AND DATALOADERS
# ==============================================================================
class ImageDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe
        self.transform = transform
        unique_labels = sorted(self.dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(unique_labels)}
    def __len__(self):
        return len(self.dataframe)
    def __getitem__(self, index):
        img_path = self.dataframe.iloc[index]['file_path']
        image = Image.open(img_path).convert("RGB")
        label_str = self.dataframe.iloc[index]['label']
        label = self.label_map[label_str]
        if self.transform:
            image = self.transform(image)
        return image, label

def get_dataloaders(size):
    normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    train_transform = transforms.Compose([
        transforms.RandomRotation(degrees=15),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomVerticalFlip(p=0.5),
        transforms.RandomResizedCrop(size, scale=(0.9, 1.0)),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.ToTensor(),
        normalize
    ])
    val_test_transform = transforms.Compose([transforms.Resize(size), transforms.ToTensor(), normalize])
    train_dataset = ImageDataset(train_dataframe, transform=train_transform)
    valid_dataset = ImageDataset(validation_dataframe, transform=val_test_transform)
    test_dataset = ImageDataset(test_dataframe, transform=val_test_transform)
    print("\n--- Label Encoding ---")
    for label, idx in train_dataset.label_map.items():
        print(f"{label} -> {idx}")
    train_dataloader = DataLoader(train_dataset, batch_size=train_batch, shuffle=True, num_workers=2)
    validation_dataloader = DataLoader(valid_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    test_dataloader = DataLoader(test_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    return train_dataloader, validation_dataloader, test_dataloader, train_dataset.label_map

train_dataloader, validation_dataloader, test_dataloader, label_map = get_dataloaders(size=image_size)
class_names = list(label_map.keys())

# ==============================================================================
# 6. DEEP LEARNING MODEL (AlexNet)
# ==============================================================================
print("\n--- Initializing AlexNet Model ---")
# Load the pre-trained AlexNet model
model = models.alexnet(weights='IMAGENET1K_V1')

# Replace the final classifier layer
num_ftrs = model.classifier[6].in_features
model.classifier[6] = nn.Linear(num_ftrs, total_class)
model = model.to(device)

optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=decay)
criterion = nn.CrossEntropyLoss()
summary(model, input_size=(train_batch, 3, 224, 224))

# ==============================================================================
# 7. MODEL TRAINING AND FINE-TUNING
# ==============================================================================
def train_finetune_model(model, criterion, optimizer, train_loader, val_loader, initial_epochs, finetune_epochs, patience, ckpt_path):
    train_loss_history, val_loss_history = [], []
    train_acc_history, val_acc_history = [], []
    best_val_acc = 0.0
    no_improvement_count = 0
    total_start_time = time.time()
    
    # --- STAGE 1: Initial Training (Feature Extraction) ---
    print("\n--- Stage 1: Training the Classifier (Frozen Base) ---")
    # Freeze all feature layers first
    for param in model.features.parameters():
        param.requires_grad = False
    
    # Ensure the classifier is trainable
    for param in model.classifier.parameters():
        param.requires_grad = True

    for epoch in range(initial_epochs):
        model.train()
        running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{initial_epochs} [T]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)
        
        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)

        model.eval()
        running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()
        
        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)
        
        print(f"Epoch {epoch+1}/{initial_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")

        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 1 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
        
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 1 after {patience} epochs with no improvement.")
            break
            
    # --- STAGE 2: Fine-Tuning ---
    print("\n--- Stage 2: Fine-Tuning (Unfrozen Base) ---")
    # Unfreeze all layers
    for param in model.parameters():
        param.requires_grad = True
    
    # Use a much lower learning rate for fine-tuning
    optimizer_ft = optim.Adam(model.parameters(), lr=learning_rate / 10)
    
    no_improvement_count = 0 # Reset counter

    for epoch in range(finetune_epochs):
        model.train()
        running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{finetune_epochs} [FT]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer_ft.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer_ft.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)

        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)
        
        model.eval()
        running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()

        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)

        print(f"Epoch {epoch+1}/{finetune_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 2 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
            print(f"   (No improvement for {no_improvement_count} epochs)")
            
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 2 after {patience} epochs with no improvement.")
            break
            
    total_end_time = time.time()
    total_training_time = total_end_time - total_start_time
    print(f"\n--- Training & Fine-Tuning Finished ---")
    print(f"Total Time: {total_training_time:.2f} seconds ({total_training_time/60:.2f} minutes)")

    return train_loss_history, train_acc_history, val_loss_history, val_acc_history

# Train and fine-tune the model
train_loss, train_acc, val_loss, val_acc = train_finetune_model(
    model, criterion, optimizer, train_dataloader, validation_dataloader, 
    initial_epochs, finetune_epochs, patience, checkpoints_path
)

# --- Plot Training History ---
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_loss, label='Train Loss')
plt.plot(val_loss, label='Validation Loss')
plt.axvline(x=initial_epochs-1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Loss over Epochs')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(train_acc, label='Train Accuracy')
plt.plot(val_acc, label='Validation Accuracy')
plt.axvline(x=initial_epochs-1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Accuracy over Epochs')
plt.legend()
plt.tight_layout()
plt.show()

# ==============================================================================
# 8. EVALUATION
# ==============================================================================
print("\n--- Evaluating Model on Test Set ---")
# Load the best model weights for final evaluation
model.load_state_dict(torch.load(os.path.join(checkpoints_path, "best_model.pt")))

def evaluate_model_full(model, dataloader):
    model.eval()
    y_true, y_pred = [], []
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, predicted = torch.max(outputs, 1)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(predicted.cpu().numpy())
    return np.array(y_true), np.array(y_pred)

y_true, y_pred = evaluate_model_full(model, test_dataloader)

# --- Overall Test Accuracy ---
test_accuracy = accuracy_score(y_true, y_pred)
print(f"Overall Test Accuracy: {test_accuracy:.3f}")

# --- Classification Report ---
print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

# --- Confusion Matrix ---
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()

# ==============================================================================
# 9. EXPLAINABLE AI (Grad-CAM)
# ==============================================================================
class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer
        self.gradients = None
        self.activations = None
        self.hook_handles = []
        self._register_hooks()

    def _register_hooks(self):
        def forward_hook(module, input, output):
            self.activations = output.detach()
        def backward_hook(module, grad_input, grad_output):
            self.gradients = grad_output[0].detach()
        self.hook_handles.extend([
            self.target_layer.register_forward_hook(forward_hook),
            self.target_layer.register_backward_hook(backward_hook)
        ])

    def remove_hooks(self):
        for handle in self.hook_handles:
            handle.remove()

    def generate_cam(self, input_tensor, target_class=None):
        self.model.eval()
        input_tensor.requires_grad_(True)
        output = self.model(input_tensor)
        
        if target_class is None:
            target_class = output.argmax(dim=1).item()
            
        self.model.zero_grad()
        class_score = output[:, target_class]
        class_score.backward()
        
        weights = torch.mean(self.gradients, dim=(2, 3), keepdim=True)
        cam = torch.sum(weights * self.activations, dim=1).squeeze().cpu().numpy()
        cam = np.maximum(cam, 0)
        cam = cv2.resize(cam, (224, 224))
        cam = (cam - np.min(cam)) / (np.max(cam) - np.min(cam) + 1e-8)
        return cam

def denormalize_image(tensor, mean, std):
    tensor = tensor.clone()
    for t, m, s in zip(tensor, mean, std):
        t.mul_(s).add_(m)
    return tensor

def show_gradcam(model, dataloader, device, target_layer, classes, num_samples=5):
    gradcam = GradCAM(model, target_layer)
    data_iter = iter(dataloader)
    
    plt.figure(figsize=(8, num_samples * 3))
    
    for i in range(num_samples):
        inputs, labels = next(data_iter)
        input_tensor = inputs[0].unsqueeze(0).to(device)
        label = labels[0].item()
        
        output = model(input_tensor)
        pred_class = output.argmax(dim=1).item()

        cam = gradcam.generate_cam(input_tensor, pred_class)
        
        mean_t = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
        std_t = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
        img_denorm = denormalize_image(inputs[0].cpu(), mean_t, std_t)
        img_np = img_denorm.numpy().transpose(1, 2, 0)
        img_np = np.clip(img_np, 0, 1)

        heatmap = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)
        overlay = cv2.addWeighted(cv2.cvtColor(np.uint8(255 * img_np), cv2.COLOR_RGB2BGR), 0.6, heatmap, 0.4, 0)
        
        plt.subplot(num_samples, 2, 2 * i + 1)
        plt.imshow(img_np)
        plt.title(f"Original\nTrue: {classes[label]}")
        plt.axis('off')

        plt.subplot(num_samples, 2, 2 * i + 2)
        plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
        plt.title(f"Grad-CAM\nPredicted: {classes[pred_class]}")
        plt.axis('off')
        
    plt.tight_layout()
    plt.show()
    gradcam.remove_hooks()

print("\n--- Grad-CAM Visualization ---")
# The last convolutional layer in AlexNet's feature extractor (`features[-1]`) is a good target.
target_layer = model.features[-1]
show_gradcam(model, test_dataloader, device, target_layer, class_names, num_samples=5)
```

```python
# 1. IMPORT LIBRARIES
# ==============================================================================
# Install torchinfo if it's not already available
!pip install torchinfo --quiet

import time
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import numpy as np
import cv2
import torch
from torch import nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
import torchvision
from torchvision import transforms
from torchinfo import summary
import torchvision.models as models
import torch.optim as optim
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import random
from PIL import Image

# ==============================================================================
# 2. CONFIGURATION
# ==============================================================================
print("--- System Information ---")
try:
    gpu_info = !nvidia-smi --query-gpu=gpu_name --format=csv
    gpu_name = gpu_info[1]
    print(f"GPU: {gpu_name}")
    device = torch.device('cuda')
except Exception as e:
    gpu_name = "No GPU available"
    print(f"GPU: {gpu_name}")
    device = torch.device('cpu')

# --- Variables ---
seed = 42
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

## IMPORTANT: Update this path to your dataset location in Kaggle ##
data_dir = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"

output_dir = "/kaggle/working/processed-dataset"
checkpoints_path = "/kaggle/working/"
os.makedirs(output_dir, exist_ok=True)

# --- Hyperparameters ---
train_batch = 32
test_batch = 16
total_class = 4
learning_rate = 0.0001
decay = 1e-4
initial_epochs = 20 # Epochs for initial training with frozen layers
finetune_epochs = 15 # Epochs for fine-tuning with unfrozen layers
patience = 5
image_size = (224, 224) # DenseNet uses 224x224 input

# ==============================================================================
# 3. EXPLORE AND SPLIT THE DATASET
# ==============================================================================
def load_split(split_dir):
    file_paths, labels = [], []
    for class_name in os.listdir(split_dir):
        class_dir = os.path.join(split_dir, class_name)
        if os.path.isdir(class_dir):
            for image_name in os.listdir(class_dir):
                file_paths.append(os.path.join(class_dir, image_name))
                labels.append(class_name)
    return pd.DataFrame({"file_path": file_paths, "label": labels})

data = load_split(data_dir)
train_dataframe, temp_dataframe = train_test_split(data, test_size=0.30, stratify=data['label'], random_state=seed)
validation_dataframe, test_dataframe = train_test_split(temp_dataframe, test_size=0.50, stratify=temp_dataframe['label'], random_state=seed)

# --- Visualize Dataset Split Ratios ---
sizes = {"Train": len(train_dataframe), "Validation": len(validation_dataframe), "Test": len(test_dataframe)}
total = sum(sizes.values())
ratios = {k: (v / total) * 100 for k, v in sizes.items()}

print("\n--- Dataset Split Ratios ---")
for split in sizes.keys():
    print(f"{split}: {sizes[split]} images ({ratios[split]:.2f}%)")

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].pie(sizes.values(), labels=sizes.keys(), autopct="%.1f%%", startangle=90, colors=["#a6cee3", "#1f78b4", "#b2df8a"])
axes[0].set_title("Dataset Split Pie Chart")
axes[1].bar(sizes.keys(), sizes.values(), color=["#a6cee3", "#1f78b4", "#b2df8a"])
axes[1].set_ylabel("Number of Images")
axes[1].set_title("Dataset Split Bar Chart")
y_max = max(sizes.values()) * 1.1
axes[1].set_ylim(0, y_max)
for i, v in enumerate(sizes.values()):
    axes[1].text(i, v + total*0.01, str(v), ha='center')  
plt.tight_layout()
plt.show()

# --- Visualize Class Distribution ---
def class_distribution_plot(train_df, test_df, val_df):
    datasets = { "Train": train_df, "Validation": val_df, "Test": test_df }
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle('Class Distribution Across Datasets', fontsize=16)
    
    for ax, (split_name, df) in zip(axes, datasets.items()):
        class_counts = df['label'].value_counts().sort_index()
        class_counts.plot(kind='bar', ax=ax, color=sns.color_palette("viridis", len(class_counts)))
        ax.set_xlabel('Classes')
        ax.set_ylabel('Number of Images')
        ax.set_title(f'{split_name} Set')
        ax.set_xticklabels(class_counts.index, rotation=45, ha='right')
        
        for i, count in enumerate(class_counts):
            ax.text(i, count + 2, str(count), ha='center')
        ax.set_ylim(0, max(class_counts) * 1.2)
    
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

class_distribution_plot(train_dataframe, test_dataframe, validation_dataframe)

# ==============================================================================
# 4. DATA PREPROCESSING (Resize and Save)
# ==============================================================================
def data_preprocess(df, split_name, size, quality=95):
    split_dir = os.path.join(output_dir, split_name)
    os.makedirs(split_dir, exist_ok=True)
    processed_file_paths = []
    for idx, row in tqdm(df.iterrows(), total=len(df), desc=f"Processing {split_name} set"):
        label = row['label']
        label_dir = os.path.join(split_dir, label)
        os.makedirs(label_dir, exist_ok=True)
        img = Image.open(row['file_path']).convert("RGB")
        img = img.resize(size, Image.Resampling.LANCZOS)
        file_name = os.path.basename(row['file_path'])
        save_path = os.path.join(label_dir, file_name)
        img.save(save_path, optimize=True, quality=quality)
        processed_file_paths.append(save_path)
    df['file_path'] = processed_file_paths
    return df

train_dataframe = data_preprocess(train_dataframe, "train", size=image_size)
validation_dataframe = data_preprocess(validation_dataframe, "validation", size=image_size)
test_dataframe = data_preprocess(test_dataframe, "test", size=image_size)

# ==============================================================================
# 5. DATASET CLASS, TRANSFORMS, AND DATALOADERS
# ==============================================================================
class ImageDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe
        self.transform = transform
        unique_labels = sorted(self.dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(unique_labels)}
    def __len__(self):
        return len(self.dataframe)
    def __getitem__(self, index):
        img_path = self.dataframe.iloc[index]['file_path']
        image = Image.open(img_path).convert("RGB")
        label_str = self.dataframe.iloc[index]['label']
        label = self.label_map[label_str]
        if self.transform:
            image = self.transform(image)
        return image, label

def get_dataloaders(size):
    normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    train_transform = transforms.Compose([
        transforms.RandomRotation(degrees=15),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomVerticalFlip(p=0.5),
        transforms.RandomResizedCrop(size, scale=(0.9, 1.0)),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.ToTensor(),
        normalize
    ])
    val_test_transform = transforms.Compose([transforms.Resize(size), transforms.ToTensor(), normalize])
    train_dataset = ImageDataset(train_dataframe, transform=train_transform)
    valid_dataset = ImageDataset(validation_dataframe, transform=val_test_transform)
    test_dataset = ImageDataset(test_dataframe, transform=val_test_transform)
    print("\n--- Label Encoding ---")
    for label, idx in train_dataset.label_map.items():
        print(f"{label} -> {idx}")
    train_dataloader = DataLoader(train_dataset, batch_size=train_batch, shuffle=True, num_workers=2)
    validation_dataloader = DataLoader(valid_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    test_dataloader = DataLoader(test_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    return train_dataloader, validation_dataloader, test_dataloader, train_dataset.label_map

train_dataloader, validation_dataloader, test_dataloader, label_map = get_dataloaders(size=image_size)
class_names = list(label_map.keys())

# ==============================================================================
# 6. DEEP LEARNING MODEL (DenseNet121)
# ==============================================================================
print("\n--- Initializing DenseNet121 Model ---")
# Load the pre-trained DenseNet121 model
model = models.densenet121(weights='IMAGENET1K_V1')

# Replace the final classifier layer
num_ftrs = model.classifier.in_features
model.classifier = nn.Linear(num_ftrs, total_class)
model = model.to(device)

optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=decay)
criterion = nn.CrossEntropyLoss()
summary(model, input_size=(train_batch, 3, 224, 224))

# ==============================================================================
# 7. MODEL TRAINING AND FINE-TUNING
# ==============================================================================
def train_finetune_model(model, criterion, optimizer, train_loader, val_loader, initial_epochs, finetune_epochs, patience, ckpt_path):
    train_loss_history, val_loss_history = [], []
    train_acc_history, val_acc_history = [], []
    best_val_acc = 0.0
    no_improvement_count = 0
    total_start_time = time.time()
    
    # --- STAGE 1: Initial Training (Feature Extraction) ---
    print("\n--- Stage 1: Training the Classifier (Frozen Base) ---")
    # Freeze all feature layers first
    for param in model.features.parameters():
        param.requires_grad = False
    
    # Ensure the classifier is trainable
    for param in model.classifier.parameters():
        param.requires_grad = True

    for epoch in range(initial_epochs):
        model.train()
        running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{initial_epochs} [T]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)
        
        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)

        model.eval()
        running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()
        
        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)
        
        print(f"Epoch {epoch+1}/{initial_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")

        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 1 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
        
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 1 after {patience} epochs with no improvement.")
            break
            
    # --- STAGE 2: Fine-Tuning ---
    print("\n--- Stage 2: Fine-Tuning (Unfrozen Base) ---")
    # Unfreeze all layers
    for param in model.parameters():
        param.requires_grad = True
    
    # Use a much lower learning rate for fine-tuning
    optimizer_ft = optim.Adam(model.parameters(), lr=learning_rate / 10)
    
    no_improvement_count = 0 # Reset counter

    for epoch in range(finetune_epochs):
        model.train()
        running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{finetune_epochs} [FT]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer_ft.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer_ft.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)

        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)
        
        model.eval()
        running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()

        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)

        print(f"Epoch {epoch+1}/{finetune_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 2 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
            print(f"   (No improvement for {no_improvement_count} epochs)")
            
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 2 after {patience} epochs with no improvement.")
            break
            
    total_end_time = time.time()
    total_training_time = total_end_time - total_start_time
    print(f"\n--- Training & Fine-Tuning Finished ---")
    print(f"Total Time: {total_training_time:.2f} seconds ({total_training_time/60:.2f} minutes)")

    return train_loss_history, train_acc_history, val_loss_history, val_acc_history

# Train and fine-tune the model
train_loss, train_acc, val_loss, val_acc = train_finetune_model(
    model, criterion, optimizer, train_dataloader, validation_dataloader, 
    initial_epochs, finetune_epochs, patience, checkpoints_path
)

# --- Plot Training History ---
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_loss, label='Train Loss')
plt.plot(val_loss, label='Validation Loss')
plt.axvline(x=initial_epochs-1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Loss over Epochs')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(train_acc, label='Train Accuracy')
plt.plot(val_acc, label='Validation Accuracy')
plt.axvline(x=initial_epochs-1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Accuracy over Epochs')
plt.legend()
plt.tight_layout()
plt.show()

# ==============================================================================
# 8. EVALUATION
# ==============================================================================
print("\n--- Evaluating Model on Test Set ---")
# Load the best model weights for final evaluation
model.load_state_dict(torch.load(os.path.join(checkpoints_path, "best_model.pt")))

def evaluate_model_full(model, dataloader):
    model.eval()
    y_true, y_pred = [], []
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, predicted = torch.max(outputs, 1)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(predicted.cpu().numpy())
    return np.array(y_true), np.array(y_pred)

y_true, y_pred = evaluate_model_full(model, test_dataloader)

# --- Overall Test Accuracy ---
test_accuracy = accuracy_score(y_true, y_pred)
print(f"Overall Test Accuracy: {test_accuracy:.3f}")

# --- Classification Report ---
print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

# --- Confusion Matrix ---
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()

# ==============================================================================
# 9. EXPLAINABLE AI (Grad-CAM)
# ==============================================================================
class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer
        self.gradients = None
        self.activations = None
        self.hook_handles = []
        self._register_hooks()

    def _register_hooks(self):
        def forward_hook(module, input, output):
            self.activations = output.detach()
        def backward_hook(module, grad_input, grad_output):
            self.gradients = grad_output[0].detach()
        self.hook_handles.extend([
            self.target_layer.register_forward_hook(forward_hook),
            self.target_layer.register_backward_hook(backward_hook)
        ])

    def remove_hooks(self):
        for handle in self.hook_handles:
            handle.remove()

    def generate_cam(self, input_tensor, target_class=None):
        self.model.eval()
        input_tensor.requires_grad_(True)
        output = self.model(input_tensor)
        
        if target_class is None:
            target_class = output.argmax(dim=1).item()
            
        self.model.zero_grad()
        class_score = output[:, target_class]
        class_score.backward()
        
        weights = torch.mean(self.gradients, dim=(2, 3), keepdim=True)
        cam = torch.sum(weights * self.activations, dim=1).squeeze().cpu().numpy()
        cam = np.maximum(cam, 0)
        cam = cv2.resize(cam, (224, 224))
        cam = (cam - np.min(cam)) / (np.max(cam) - np.min(cam) + 1e-8)
        return cam

def denormalize_image(tensor, mean, std):
    tensor = tensor.clone()
    for t, m, s in zip(tensor, mean, std):
        t.mul_(s).add_(m)
    return tensor

def show_gradcam(model, dataloader, device, target_layer, classes, num_samples=5):
    gradcam = GradCAM(model, target_layer)
    data_iter = iter(dataloader)
    
    plt.figure(figsize=(8, num_samples * 3))
    
    for i in range(num_samples):
        inputs, labels = next(data_iter)
        input_tensor = inputs[0].unsqueeze(0).to(device)
        label = labels[0].item()
        
        output = model(input_tensor)
        pred_class = output.argmax(dim=1).item()

        cam = gradcam.generate_cam(input_tensor, pred_class)
        
        mean_t = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
        std_t = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
        img_denorm = denormalize_image(inputs[0].cpu(), mean_t, std_t)
        img_np = img_denorm.numpy().transpose(1, 2, 0)
        img_np = np.clip(img_np, 0, 1)

        heatmap = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)
        overlay = cv2.addWeighted(cv2.cvtColor(np.uint8(255 * img_np), cv2.COLOR_RGB2BGR), 0.6, heatmap, 0.4, 0)
        
        plt.subplot(num_samples, 2, 2 * i + 1)
        plt.imshow(img_np)
        plt.title(f"Original\nTrue: {classes[label]}")
        plt.axis('off')

        plt.subplot(num_samples, 2, 2 * i + 2)
        plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
        plt.title(f"Grad-CAM\nPredicted: {classes[pred_class]}")
        plt.axis('off')
        
    plt.tight_layout()
    plt.show()
    gradcam.remove_hooks()

print("\n--- Grad-CAM Visualization ---")
# The last dense block in DenseNet121 (`features.denseblock4`) is a good target for Grad-CAM.
# We will hook into the final convolutional layer of that block.
target_layer = model.features.denseblock4.denselayer16.conv2
show_gradcam(model, test_dataloader, device, target_layer, class_names, num_samples=5)
```

```python
# 1. IMPORT LIBRARIES
# ==============================================================================
# Install torchinfo if it's not already available
!pip install torchinfo --quiet

import time
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_curve, auc
from sklearn.preprocessing import label_binarize
import numpy as np
import cv2
import torch
from torch import nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
import torchvision
from torchvision import transforms
from torchinfo import summary
import torchvision.models as models
import torch.optim as optim
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import random
from PIL import Image
from itertools import cycle

# ==============================================================================
# 2. CONFIGURATION
# ==============================================================================
print("--- System Information ---")
try:
    gpu_info = !nvidia-smi --query-gpu=gpu_name --format=csv
    gpu_name = gpu_info[1]
    print(f"GPU: {gpu_name}")
    device = torch.device('cuda')
except Exception as e:
    gpu_name = "No GPU available"
    print(f"GPU: {gpu_name}")
    device = torch.device('cpu')

# --- Variables ---
seed = 42
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

## IMPORTANT: Update this path to your dataset location in Kaggle ##
data_dir = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"

output_dir = "/kaggle/working/processed-dataset"
checkpoints_path = "/kaggle/working/"
os.makedirs(output_dir, exist_ok=True)

# --- Hyperparameters ---
train_batch = 32
test_batch = 16
total_class = 4
learning_rate = 0.0001
decay = 1e-4
initial_epochs = 20 # Epochs for initial training with frozen layers
finetune_epochs = 15 # Epochs for fine-tuning
patience = 5
image_size = (224, 224) # EfficientNetB0 uses 224x224

# ==============================================================================
# 3. EXPLORE AND SPLIT THE DATASET
# ==============================================================================
def load_split(split_dir):
    file_paths, labels = [], []
    for class_name in os.listdir(split_dir):
        class_dir = os.path.join(split_dir, class_name)
        if os.path.isdir(class_dir):
            for image_name in os.listdir(class_dir):
                file_paths.append(os.path.join(class_dir, image_name))
                labels.append(class_name)
    return pd.DataFrame({"file_path": file_paths, "label": labels})

data = load_split(data_dir)
train_dataframe, temp_dataframe = train_test_split(data, test_size=0.30, stratify=data['label'], random_state=seed)
validation_dataframe, test_dataframe = train_test_split(temp_dataframe, test_size=0.50, stratify=temp_dataframe['label'], random_state=seed)

# --- Visualize Dataset Split Ratios ---
sizes = {
    "Train": len(train_dataframe),
    "Validation": len(validation_dataframe),
    "Test": len(test_dataframe),
}

total = sum(sizes.values())
ratios = {k: (v / total) * 100 for k, v in sizes.items()}

print("\n--- Dataset Split Ratios ---")
for split in sizes.keys():
    print(f"{split}: {sizes[split]} images ({ratios[split]:.2f}%)")

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].pie(
    sizes.values(),
    labels=sizes.keys(),
    autopct="%.1f%%",
    startangle=90,
    colors=["#a6cee3", "#1f78b4", "#b2df8a"]
)
axes[0].set_title("Dataset Split Pie Chart")

axes[1].bar(sizes.keys(), sizes.values(), color=["#a6cee3", "#1f78b4", "#b2df8a"])
axes[1].set_ylabel("Number of Images")
axes[1].set_title("Dataset Split Bar Chart")
y_max = max(sizes.values()) * 1.1
axes[1].set_ylim(0, y_max)
for i, v in enumerate(sizes.values()):
    axes[1].text(i, v + total*0.01, str(v), ha='center')  
plt.tight_layout()
plt.show()

# --- Visualize Class Distribution ---
def class_distribution_plot(train_df, test_df, val_df):
    datasets = { "Train": train_df, "Validation": val_df, "Test": test_df }
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle('Class Distribution Across Datasets', fontsize=16)
    
    for ax, (split_name, df) in zip(axes, datasets.items()):
        class_counts = df['label'].value_counts().sort_index()
        class_counts.plot(kind='bar', ax=ax, color=sns.color_palette("viridis", len(class_counts)))
        ax.set_xlabel('Classes')
        ax.set_ylabel('Number of Images')
        ax.set_title(f'{split_name} Set')
        ax.set_xticklabels(class_counts.index, rotation=45, ha='right')
        
        for i, count in enumerate(class_counts):
            ax.text(i, count + 2, str(count), ha='center')
        ax.set_ylim(0, max(class_counts) * 1.2)
    
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

class_distribution_plot(train_dataframe, test_dataframe, validation_dataframe)

# ==============================================================================
# 4. DATA PREPROCESSING (Resize and Save)
# ==============================================================================
def data_preprocess(df, split_name, size, quality=95):
    split_dir = os.path.join(output_dir, split_name)
    os.makedirs(split_dir, exist_ok=True)
    processed_file_paths = []
    for idx, row in tqdm(df.iterrows(), total=len(df), desc=f"Processing {split_name} set"):
        label = row['label']
        label_dir = os.path.join(split_dir, label)
        os.makedirs(label_dir, exist_ok=True)
        img = Image.open(row['file_path']).convert("RGB").resize(size, Image.Resampling.LANCZOS)
        file_name = os.path.basename(row['file_path'])
        save_path = os.path.join(label_dir, file_name)
        img.save(save_path, optimize=True, quality=quality)
        processed_file_paths.append(save_path)
    df['file_path'] = processed_file_paths
    return df

train_dataframe = data_preprocess(train_dataframe, "train", size=image_size)
validation_dataframe = data_preprocess(validation_dataframe, "validation", size=image_size)
test_dataframe = data_preprocess(test_dataframe, "test", size=image_size)

# ==============================================================================
# 5. DATASET CLASS, TRANSFORMS, AND DATALOADERS
# ==============================================================================
class ImageDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe
        self.transform = transform
        unique_labels = sorted(self.dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(unique_labels)}
    def __len__(self): return len(self.dataframe)
    def __getitem__(self, index):
        img_path = self.dataframe.iloc[index]['file_path']
        image = Image.open(img_path).convert("RGB")
        label = self.label_map[self.dataframe.iloc[index]['label']]
        if self.transform: image = self.transform(image)
        return image, label

def get_dataloaders(size):
    normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    train_transform = transforms.Compose([
        transforms.RandomRotation(degrees=15), transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomVerticalFlip(p=0.5), transforms.RandomResizedCrop(size, scale=(0.9, 1.0)),
        transforms.ColorJitter(brightness=0.2, contrast=0.2), transforms.ToTensor(), normalize])
    val_test_transform = transforms.Compose([transforms.Resize(size), transforms.ToTensor(), normalize])
    train_dataset = ImageDataset(train_dataframe, transform=train_transform)
    valid_dataset = ImageDataset(validation_dataframe, transform=val_test_transform)
    test_dataset = ImageDataset(test_dataframe, transform=val_test_transform)
    print("\n--- Label Encoding ---")
    for label, idx in train_dataset.label_map.items(): print(f"{label} -> {idx}")
    train_dataloader = DataLoader(train_dataset, batch_size=train_batch, shuffle=True, num_workers=2)
    validation_dataloader = DataLoader(valid_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    test_dataloader = DataLoader(test_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    return train_dataloader, validation_dataloader, test_dataloader, train_dataset.label_map

train_dataloader, validation_dataloader, test_dataloader, label_map = get_dataloaders(size=image_size)
class_names = list(label_map.keys())

# ==============================================================================
# 6. DEEP LEARNING MODEL (EfficientNetB0)
# ==============================================================================
print("\n--- Initializing EfficientNetB0 Model ---")
model = models.efficientnet_b0(weights='IMAGENET1K_V1')
num_ftrs = model.classifier[1].in_features
model.classifier[1] = nn.Linear(num_ftrs, total_class)
model = model.to(device)
optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=decay)
criterion = nn.CrossEntropyLoss()
summary(model, input_size=(train_batch, 3, 224, 224))

# ==============================================================================
# 7. MODEL TRAINING AND FINE-TUNING
# ==============================================================================
def train_finetune_model(model, criterion, optimizer, train_loader, val_loader, initial_epochs, finetune_epochs, patience, ckpt_path):
    train_loss_history, val_loss_history = [], []
    train_acc_history, val_acc_history = [], []
    best_val_acc = 0.0
    no_improvement_count = 0
    total_start_time = time.time()
    
    # --- STAGE 1: Initial Training (Feature Extraction) ---
    print("\n--- Stage 1: Training the Classifier (Frozen Base) ---")
    for param in model.features.parameters(): param.requires_grad = False
    for param in model.classifier.parameters(): param.requires_grad = True
    for epoch in range(initial_epochs):
        model.train(); running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{initial_epochs} [T]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad(); outputs = model(inputs); loss = criterion(outputs, labels); loss.backward(); optimizer.step()
            running_loss += loss.item() * inputs.size(0); _, predicted = torch.max(outputs, 1); total_train += labels.size(0); correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)
        
        epoch_train_loss = running_loss / total_train; epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss); train_acc_history.append(epoch_train_acc)
        
        model.eval(); running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs); loss = criterion(outputs, labels); running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1); total_val += labels.size(0); correct_val += (predicted == labels).sum().item()
        
        epoch_val_loss = running_val_loss / total_val; epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss); val_acc_history.append(epoch_val_acc)
        print(f"Epoch {epoch+1}/{initial_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 1 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else: no_improvement_count += 1
        if no_improvement_count >= patience: print(f"🛑 Early stopping Stage 1 after {patience} epochs with no improvement."); break
            
    # --- STAGE 2: Fine-Tuning ---
    print("\n--- Stage 2: Fine-Tuning (Unfrozen Base) ---")
    for param in model.parameters(): param.requires_grad = True
    optimizer_ft = optim.Adam(model.parameters(), lr=learning_rate / 10)
    no_improvement_count = 0
    for epoch in range(finetune_epochs):
        model.train(); running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{finetune_epochs} [FT]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer_ft.zero_grad(); outputs = model(inputs); loss = criterion(outputs, labels); loss.backward(); optimizer_ft.step()
            running_loss += loss.item() * inputs.size(0); _, predicted = torch.max(outputs, 1); total_train += labels.size(0); correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)

        epoch_train_loss = running_loss / total_train; epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss); train_acc_history.append(epoch_train_acc)
        
        model.eval(); running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs); loss = criterion(outputs, labels); running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1); total_val += labels.size(0); correct_val += (predicted == labels).sum().item()

        epoch_val_loss = running_val_loss / total_val; epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss); val_acc_history.append(epoch_val_acc)
        print(f"Epoch {epoch+1}/{finetune_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 2 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
            print(f"   (No improvement for {no_improvement_count} epochs)")
        if no_improvement_count >= patience: print(f"🛑 Early stopping Stage 2 after {patience} epochs with no improvement."); break
            
    total_end_time = time.time()
    print(f"\n--- Training & Fine-Tuning Finished ---")
    print(f"Total Time: {(total_end_time - total_start_time)/60:.2f} minutes")
    return train_loss_history, train_acc_history, val_loss_history, val_acc_history

# Train and fine-tune the model
train_loss, train_acc, val_loss, val_acc = train_finetune_model(
    model, criterion, optimizer, train_dataloader, validation_dataloader, 
    initial_epochs, finetune_epochs, patience, checkpoints_path
)

# --- Plot Training History ---
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_loss, label='Train Loss'); plt.plot(val_loss, label='Validation Loss')
plt.axvline(x=len(train_loss) - finetune_epochs - 1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs'); plt.ylabel('Loss'); plt.title('Loss over Epochs'); plt.legend()
plt.subplot(1, 2, 2)
plt.plot(train_acc, label='Train Accuracy'); plt.plot(val_acc, label='Validation Accuracy')
plt.axvline(x=len(train_acc) - finetune_epochs - 1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs'); plt.ylabel('Accuracy'); plt.title('Accuracy over Epochs'); plt.legend()
plt.tight_layout(); plt.show()

# ==============================================================================
# 8. EVALUATION
# ==============================================================================
print("\n--- Evaluating Model on Test Set ---")
model.load_state_dict(torch.load(os.path.join(checkpoints_path, "best_model.pt")))
def evaluate_model_full(model, dataloader):
    model.eval(); y_true, y_pred, y_prob = [], [], []
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            probs = F.softmax(outputs, dim=1)
            _, predicted = torch.max(outputs, 1)
            y_true.extend(labels.cpu().numpy()); y_pred.extend(predicted.cpu().numpy()); y_prob.extend(probs.cpu().numpy())
    return np.array(y_true), np.array(y_pred), np.array(y_prob)

y_true, y_pred, y_prob = evaluate_model_full(model, test_dataloader)
print(f"Overall Test Accuracy: {accuracy_score(y_true, y_pred):.3f}")
print("\nClassification Report:"); print(classification_report(y_true, y_pred, target_names=class_names))
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label'); plt.ylabel('True Label'); plt.title('Confusion Matrix'); plt.show()

# --- ROC Curve and AUC Score ---
y_true_binarized = label_binarize(y_true, classes=range(total_class))
fpr, tpr, roc_auc = {}, {}, {}
for i in range(total_class):
    fpr[i], tpr[i], _ = roc_curve(y_true_binarized[:, i], y_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])
plt.figure(figsize=(10, 8)); colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green'])
for i, color in zip(range(total_class), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2, label=f'ROC curve of class {class_names[i]} (AUC = {roc_auc[i]:0.2f})')
plt.plot([0, 1], [0, 1], 'k--', lw=2); plt.xlim([0.0, 1.0]); plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.title('Multi-class ROC Curve for EfficientNetB0'); plt.legend(loc="lower right"); plt.show()

# ==============================================================================
# 9. EXPLAINABLE AI (Grad-CAM)
# ==============================================================================
class GradCAM:
    def __init__(self, model, target_layer):
        self.model=model; self.target_layer=target_layer; self.gradients=None; self.activations=None; self.hook_handles=[]
        self._register_hooks()
    def _register_hooks(self):
        def f_hook(module, input, output): self.activations = output.detach()
        def b_hook(module, grad_input, grad_output): self.gradients = grad_output[0].detach()
        self.hook_handles.extend([self.target_layer.register_forward_hook(f_hook), self.target_layer.register_backward_hook(b_hook)])
    def remove_hooks(self):
        for handle in self.hook_handles: handle.remove()
    def generate_cam(self, input_tensor, target_class=None):
        self.model.eval(); input_tensor.requires_grad_(True); output = self.model(input_tensor)
        if target_class is None: target_class = output.argmax(dim=1).item()
        self.model.zero_grad(); class_score = output[:, target_class]; class_score.backward()
        weights = torch.mean(self.gradients, dim=(2, 3), keepdim=True)
        cam = torch.sum(weights * self.activations, dim=1).squeeze().cpu().numpy()
        cam = np.maximum(cam, 0); cam = cv2.resize(cam, (224, 224)); cam = (cam - np.min(cam)) / (np.max(cam) - np.min(cam) + 1e-8)
        return cam
def denormalize_image(tensor, mean, std):
    tensor = tensor.clone()
    for t, m, s in zip(tensor, mean, std): t.mul_(s).add_(m)
    return tensor
def show_gradcam(model, dataloader, device, target_layer, classes, num_samples=5):
    gradcam = GradCAM(model, target_layer); data_iter = iter(dataloader)
    plt.figure(figsize=(8, num_samples * 3))
    for i in range(num_samples):
        inputs, labels = next(data_iter); input_tensor = inputs[0].unsqueeze(0).to(device); label = labels[0].item()
        output = model(input_tensor); pred_class = output.argmax(dim=1).item()
        cam = gradcam.generate_cam(input_tensor, pred_class)
        mean_t = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1); std_t = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
        img_denorm = denormalize_image(inputs[0].cpu(), mean_t, std_t); img_np = np.clip(img_denorm.numpy().transpose(1, 2, 0), 0, 1)
        heatmap = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)
        overlay = cv2.addWeighted(cv2.cvtColor(np.uint8(255 * img_np), cv2.COLOR_RGB2BGR), 0.6, heatmap, 0.4, 0)
        plt.subplot(num_samples, 2, 2 * i + 1); plt.imshow(img_np); plt.title(f"Original\nTrue: {classes[label]}"); plt.axis('off')
        plt.subplot(num_samples, 2, 2 * i + 2); plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB)); plt.title(f"Grad-CAM\nPredicted: {classes[pred_class]}"); plt.axis('off')
    plt.tight_layout(); plt.show(); gradcam.remove_hooks()

print("\n--- Grad-CAM Visualization ---")
target_layer = model.features[-1][0]
show_gradcam(model, test_dataloader, device, target_layer, class_names, num_samples=5)
```

```python
# 1. IMPORT LIBRARIES
# ==============================================================================
# Install torchinfo if it's not already available
!pip install torchinfo --quiet

import time
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_curve, auc
from sklearn.preprocessing import label_binarize
import numpy as np
import cv2
import torch
from torch import nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset, random_split
import torchvision
from torchvision import transforms
from torchinfo import summary
import torchvision.models as models
import torch.optim as optim
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import random
from PIL import Image
from itertools import cycle

# ==============================================================================
# 2. CONFIGURATION
# ==============================================================================
print("--- System Information ---")
try:
    gpu_info = !nvidia-smi --query-gpu=gpu_name --format=csv
    gpu_name = gpu_info[1]
    print(f"GPU: {gpu_name}")
    device = torch.device('cuda')
except Exception as e:
    gpu_name = "No GPU available"
    print(f"GPU: {gpu_name}")
    device = torch.device('cpu')

# --- Variables ---
seed = 42
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

## IMPORTANT: Update this path to your dataset location in Kaggle ##
data_dir = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"

output_dir = "/kaggle/working/processed-dataset"
checkpoints_path = "/kaggle/working/"
os.makedirs(output_dir, exist_ok=True)

# --- Hyperparameters ---
train_batch = 32
test_batch = 16
total_class = 4
learning_rate = 0.0001
decay = 1e-4
initial_epochs = 20
finetune_epochs = 15
patience = 5
image_size = (224, 224)

# ==============================================================================
# 3. EXPLORE AND SPLIT THE DATASET
# ==============================================================================
def load_split(split_dir):
    file_paths, labels = [], []
    for class_name in os.listdir(split_dir):
        class_dir = os.path.join(split_dir, class_name)
        if os.path.isdir(class_dir):
            for image_name in os.listdir(class_dir):
                file_paths.append(os.path.join(class_dir, image_name))
                labels.append(class_name)
    return pd.DataFrame({"file_path": file_paths, "label": labels})

data = load_split(data_dir)
train_dataframe, temp_dataframe = train_test_split(data, test_size=0.30, stratify=data['label'], random_state=seed)
validation_dataframe, test_dataframe = train_test_split(temp_dataframe, test_size=0.50, stratify=temp_dataframe['label'], random_state=seed)

# --- Visualize Dataset Split Ratios ---
sizes = {"Train": len(train_dataframe), "Validation": len(validation_dataframe), "Test": len(test_dataframe)}
total = sum(sizes.values())
ratios = {k: (v / total) * 100 for k, v in sizes.items()}
print("\n--- Dataset Split Ratios ---")
for split in sizes.keys(): print(f"{split}: {sizes[split]} images ({ratios[split]:.2f}%)")
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].pie(sizes.values(), labels=sizes.keys(), autopct="%.1f%%", startangle=90, colors=["#a6cee3", "#1f78b4", "#b2df8a"])
axes[0].set_title("Dataset Split Pie Chart")
axes[1].bar(sizes.keys(), sizes.values(), color=["#a6cee3", "#1f78b4", "#b2df8a"])
axes[1].set_ylabel("Number of Images")
axes[1].set_title("Dataset Split Bar Chart")
y_max = max(sizes.values()) * 1.1
axes[1].set_ylim(0, y_max)
for i, v in enumerate(sizes.values()): axes[1].text(i, v + total*0.01, str(v), ha='center')  
plt.tight_layout()
plt.show()

# --- Visualize Class Distribution ---
def class_distribution_plot(train_df, test_df, val_df):
    datasets = { "Train": train_df, "Validation": val_df, "Test": test_df }
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle('Class Distribution Across Datasets', fontsize=16)
    for ax, (split_name, df) in zip(axes, datasets.items()):
        class_counts = df['label'].value_counts().sort_index()
        class_counts.plot(kind='bar', ax=ax, color=sns.color_palette("viridis", len(class_counts)))
        ax.set_xlabel('Classes'); ax.set_ylabel('Number of Images'); ax.set_title(f'{split_name} Set')
        ax.set_xticklabels(class_counts.index, rotation=45, ha='right')
        for i, count in enumerate(class_counts): ax.text(i, count + 2, str(count), ha='center')
        ax.set_ylim(0, max(class_counts) * 1.2)
    plt.tight_layout(rect=[0, 0, 1, 0.96]); plt.show()
class_distribution_plot(train_dataframe, test_dataframe, validation_dataframe)

# ==============================================================================
# 4. DATA PREPROCESSING (Resize and Save)
# ==============================================================================
def data_preprocess(df, split_name, size, quality=95):
    split_dir = os.path.join(output_dir, split_name)
    os.makedirs(split_dir, exist_ok=True)
    processed_file_paths = []
    for idx, row in tqdm(df.iterrows(), total=len(df), desc=f"Processing {split_name} set"):
        label = row['label']
        label_dir = os.path.join(split_dir, label)
        os.makedirs(label_dir, exist_ok=True)
        img = Image.open(row['file_path']).convert("RGB").resize(size, Image.Resampling.LANCZOS)
        file_name = os.path.basename(row['file_path'])
        save_path = os.path.join(label_dir, file_name)
        img.save(save_path, optimize=True, quality=quality)
        processed_file_paths.append(save_path)
    df['file_path'] = processed_file_paths
    return df

train_dataframe = data_preprocess(train_dataframe, "train", size=image_size)
validation_dataframe = data_preprocess(validation_dataframe, "validation", size=image_size)
test_dataframe = data_preprocess(test_dataframe, "test", size=image_size)

# ==============================================================================
# 5. DATASET CLASS, TRANSFORMS, AND DATALOADERS
# ==============================================================================
class ImageDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe
        self.transform = transform
        unique_labels = sorted(self.dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(unique_labels)}
    def __len__(self):
        return len(self.dataframe)
    def __getitem__(self, index):
        img_path = self.dataframe.iloc[index]['file_path']
        image = Image.open(img_path).convert("RGB")
        label = self.label_map[self.dataframe.iloc[index]['label']]
        if self.transform:
            image = self.transform(image)
        return image, label

def get_dataloaders(size):
    normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    train_transform = transforms.Compose([
        transforms.RandomRotation(degrees=15),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomVerticalFlip(p=0.5),
        transforms.RandomResizedCrop(size, scale=(0.9, 1.0)),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.ToTensor(),
        normalize
    ])
    val_test_transform = transforms.Compose([transforms.Resize(size), transforms.ToTensor(), normalize])
    train_dataset = ImageDataset(train_dataframe, transform=train_transform)
    valid_dataset = ImageDataset(validation_dataframe, transform=val_test_transform)
    test_dataset = ImageDataset(test_dataframe, transform=val_test_transform)
    print("\n--- Label Encoding ---")
    for label, idx in train_dataset.label_map.items(): print(f"{label} -> {idx}")
    train_dataloader = DataLoader(train_dataset, batch_size=train_batch, shuffle=True, num_workers=2)
    validation_dataloader = DataLoader(valid_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    test_dataloader = DataLoader(test_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    return train_dataloader, validation_dataloader, test_dataloader, train_dataset.label_map

train_dataloader, validation_dataloader, test_dataloader, label_map = get_dataloaders(size=image_size)
class_names = list(label_map.keys())

# ==============================================================================
# 6. DEEP LEARNING MODEL (VGG16)
# ==============================================================================
print("\n--- Initializing VGG16 Model ---")
# Use pretrained=True for compatibility with older torchvision versions
model = models.vgg16(pretrained=True)
num_ftrs = model.classifier[6].in_features
model.classifier[6] = nn.Linear(num_ftrs, total_class)
model = model.to(device)

optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=decay)
criterion = nn.CrossEntropyLoss()
summary(model, input_size=(train_batch, 3, 224, 224))

# ==============================================================================
# 7. MODEL TRAINING AND FINE-TUNING
# ==============================================================================
def train_finetune_model(model, criterion, optimizer, train_loader, val_loader, initial_epochs, finetune_epochs, patience, ckpt_path):
    train_loss_history, val_loss_history = [], []
    train_acc_history, val_acc_history = [], []
    best_val_acc = 0.0
    no_improvement_count = 0
    total_start_time = time.time()
    
    # --- STAGE 1: Initial Training (Feature Extraction) ---
    print("\n--- Stage 1: Training the Classifier (Frozen Base) ---")
    for param in model.features.parameters(): param.requires_grad = False
    for param in model.classifier.parameters(): param.requires_grad = True

    for epoch in range(initial_epochs):
        model.train()
        running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{initial_epochs} [T]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)
        
        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)

        model.eval()
        running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()
        
        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)
        
        print(f"Epoch {epoch+1}/{initial_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 1 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 1 after {patience} epochs with no improvement.")
            break
            
    # --- STAGE 2: Fine-Tuning ---
    print("\n--- Stage 2: Fine-Tuning (Unfrozen Base) ---")
    for param in model.parameters(): param.requires_grad = True
    optimizer_ft = optim.Adam(model.parameters(), lr=learning_rate / 10)
    no_improvement_count = 0 
    for epoch in range(finetune_epochs):
        model.train()
        running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{finetune_epochs} [FT]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer_ft.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer_ft.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)

        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)
        
        model.eval()
        running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()
        
        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)
        print(f"Epoch {epoch+1}/{finetune_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 2 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
            print(f"   (No improvement for {no_improvement_count} epochs)")
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 2 after {patience} epochs with no improvement.")
            break
            
    total_end_time = time.time()
    print(f"\n--- Training & Fine-Tuning Finished ---")
    print(f"Total Time: {(total_end_time - total_start_time)/60:.2f} minutes")
    return train_loss_history, train_acc_history, val_loss_history, val_acc_history

train_loss, train_acc, val_loss, val_acc = train_finetune_model(
    model, criterion, optimizer, train_dataloader, validation_dataloader, 
    initial_epochs, finetune_epochs, patience, checkpoints_path
)

# --- Plot Training History ---
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_loss, label='Train Loss'); plt.plot(val_loss, label='Validation Loss')
plt.axvline(x=initial_epochs-1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs'); plt.ylabel('Loss'); plt.title('Loss over Epochs'); plt.legend()
plt.subplot(1, 2, 2)
plt.plot(train_acc, label='Train Accuracy'); plt.plot(val_acc, label='Validation Accuracy')
plt.axvline(x=initial_epochs-1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs'); plt.ylabel('Accuracy'); plt.title('Accuracy over Epochs'); plt.legend()
plt.tight_layout(); plt.show()

# ==============================================================================
# 8. EVALUATION
# ==============================================================================
print("\n--- Evaluating Model on Test Set ---")
model.load_state_dict(torch.load(os.path.join(checkpoints_path, "best_model.pt")))
def evaluate_model_full(model, dataloader):
    model.eval(); y_true, y_pred, y_prob = [], [], []
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            probs = F.softmax(outputs, dim=1)
            _, predicted = torch.max(outputs, 1)
            y_true.extend(labels.cpu().numpy()); y_pred.extend(predicted.cpu().numpy()); y_prob.extend(probs.cpu().numpy())
    return np.array(y_true), np.array(y_pred), np.array(y_prob)
y_true, y_pred, y_prob = evaluate_model_full(model, test_dataloader)
print(f"Overall Test Accuracy: {accuracy_score(y_true, y_pred):.3f}")
print("\nClassification Report:"); print(classification_report(y_true, y_pred, target_names=class_names))
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label'); plt.ylabel('True Label'); plt.title('Confusion Matrix'); plt.show()

# --- ROC Curve and AUC Score ---
y_true_binarized = label_binarize(y_true, classes=range(total_class))
fpr, tpr, roc_auc = {}, {}, {}
for i in range(total_class):
    fpr[i], tpr[i], _ = roc_curve(y_true_binarized[:, i], y_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])
fpr["micro"], tpr["micro"], _ = roc_curve(y_true_binarized.ravel(), y_prob.ravel())
roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])
plt.figure(figsize=(10, 8)); colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green'])
for i, color in zip(range(total_class), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2, label=f'ROC curve of class {class_names[i]} (area = {roc_auc[i]:0.2f})')
plt.plot([0, 1], [0, 1], 'k--', lw=2); plt.xlim([0.0, 1.0]); plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.title('Multi-class ROC Curve'); plt.legend(loc="lower right"); plt.show()

# ==============================================================================
# 9. EXPLAINABLE AI (Grad-CAM)
# ==============================================================================
class GradCAM:
    def __init__(self, model, target_layer):
        self.model=model; self.target_layer=target_layer; self.gradients=None; self.activations=None; self.hook_handles=[]
        self._register_hooks()
    def _register_hooks(self):
        def f_hook(module, input, output): self.activations = output.detach()
        def b_hook(module, grad_input, grad_output): self.gradients = grad_output[0].detach()
        self.hook_handles.extend([self.target_layer.register_forward_hook(f_hook), self.target_layer.register_backward_hook(b_hook)])
    def remove_hooks(self):
        for handle in self.hook_handles: handle.remove()
    def generate_cam(self, input_tensor, target_class=None):
        self.model.eval(); input_tensor.requires_grad_(True); output = self.model(input_tensor)
        if target_class is None: target_class = output.argmax(dim=1).item()
        self.model.zero_grad(); class_score = output[:, target_class]; class_score.backward()
        weights = torch.mean(self.gradients, dim=(2, 3), keepdim=True)
        cam = torch.sum(weights * self.activations, dim=1).squeeze().cpu().numpy()
        cam = np.maximum(cam, 0); cam = cv2.resize(cam, (224, 224)); cam = (cam - np.min(cam)) / (np.max(cam) - np.min(cam) + 1e-8)
        return cam
def denormalize_image(tensor, mean, std):
    tensor = tensor.clone()
    for t, m, s in zip(tensor, mean, std): t.mul_(s).add_(m)
    return tensor
def show_gradcam(model, dataloader, device, target_layer, classes, num_samples=5):
    gradcam = GradCAM(model, target_layer); data_iter = iter(dataloader)
    plt.figure(figsize=(8, num_samples * 3))
    for i in range(num_samples):
        inputs, labels = next(data_iter); input_tensor = inputs[0].unsqueeze(0).to(device); label = labels[0].item()
        output = model(input_tensor); pred_class = output.argmax(dim=1).item()
        cam = gradcam.generate_cam(input_tensor, pred_class)
        mean_t = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1); std_t = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
        img_denorm = denormalize_image(inputs[0].cpu(), mean_t, std_t); img_np = np.clip(img_denorm.numpy().transpose(1, 2, 0), 0, 1)
        heatmap = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)
        overlay = cv2.addWeighted(cv2.cvtColor(np.uint8(255 * img_np), cv2.COLOR_RGB2BGR), 0.6, heatmap, 0.4, 0)
        plt.subplot(num_samples, 2, 2 * i + 1); plt.imshow(img_np); plt.title(f"Original\nTrue: {classes[label]}"); plt.axis('off')
        plt.subplot(num_samples, 2, 2 * i + 2); plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB)); plt.title(f"Grad-CAM\nPredicted: {classes[pred_class]}"); plt.axis('off')
    plt.tight_layout(); plt.show(); gradcam.remove_hooks()

print("\n--- Grad-CAM Visualization ---")
target_layer = model.features[-1]
show_gradcam(model, test_dataloader, device, target_layer, class_names, num_samples=5)
```

```python
# 1. IMPORT LIBRARIES
# ==============================================================================
# Install torchinfo if it's not already available
!pip install torchinfo --quiet

import time
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import numpy as np
import cv2
import torch
from torch import nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
import torchvision
from torchvision import transforms
from torchinfo import summary
import torchvision.models as models
import torch.optim as optim
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import random
from PIL import Image

# ==============================================================================
# 2. CONFIGURATION
# ==============================================================================
print("--- System Information ---")
try:
    gpu_info = !nvidia-smi --query-gpu=gpu_name --format=csv
    gpu_name = gpu_info[1]
    print(f"GPU: {gpu_name}")
    device = torch.device('cuda')
except Exception as e:
    gpu_name = "No GPU available"
    print(f"GPU: {gpu_name}")
    device = torch.device('cpu')

# --- Variables ---
seed = 42
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

## IMPORTANT: Update this path to your dataset location in Kaggle ##
data_dir = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"

output_dir = "/kaggle/working/processed-dataset"
checkpoints_path = "/kaggle/working/"
os.makedirs(output_dir, exist_ok=True)

# --- Hyperparameters ---
train_batch = 32
test_batch = 16
total_class = 4
learning_rate = 0.0001
decay = 1e-4
initial_epochs = 20 # Epochs for initial training with frozen layers
finetune_epochs = 15 # Epochs for fine-tuning with unfrozen layers
patience = 5
image_size = (299, 299) # Xception, like Inception, requires 299x299

# ==============================================================================
# 3. EXPLORE AND SPLIT THE DATASET
# ==============================================================================
def load_split(split_dir):
    file_paths, labels = [], []
    for class_name in os.listdir(split_dir):
        class_dir = os.path.join(split_dir, class_name)
        if os.path.isdir(class_dir):
            for image_name in os.listdir(class_dir):
                file_paths.append(os.path.join(class_dir, image_name))
                labels.append(class_name)
    return pd.DataFrame({"file_path": file_paths, "label": labels})

data = load_split(data_dir)
train_dataframe, temp_dataframe = train_test_split(data, test_size=0.30, stratify=data['label'], random_state=seed)
validation_dataframe, test_dataframe = train_test_split(temp_dataframe, test_size=0.50, stratify=temp_dataframe['label'], random_state=seed)

# --- Visualize Dataset Split Ratios ---
sizes = {"Train": len(train_dataframe), "Validation": len(validation_dataframe), "Test": len(test_dataframe)}
total = sum(sizes.values())
ratios = {k: (v / total) * 100 for k, v in sizes.items()}

print("\n--- Dataset Split Ratios ---")
for split in sizes.keys():
    print(f"{split}: {sizes[split]} images ({ratios[split]:.2f}%)")

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].pie(sizes.values(), labels=sizes.keys(), autopct="%.1f%%", startangle=90, colors=["#a6cee3", "#1f78b4", "#b2df8a"])
axes[0].set_title("Dataset Split Pie Chart")
axes[1].bar(sizes.keys(), sizes.values(), color=["#a6cee3", "#1f78b4", "#b2df8a"])
axes[1].set_ylabel("Number of Images")
axes[1].set_title("Dataset Split Bar Chart")
y_max = max(sizes.values()) * 1.1
axes[1].set_ylim(0, y_max)
for i, v in enumerate(sizes.values()):
    axes[1].text(i, v + total*0.01, str(v), ha='center')  
plt.tight_layout()
plt.show()

# --- Visualize Class Distribution ---
def class_distribution_plot(train_df, test_df, val_df):
    datasets = { "Train": train_df, "Validation": val_df, "Test": test_df }
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle('Class Distribution Across Datasets', fontsize=16)
    
    for ax, (split_name, df) in zip(axes, datasets.items()):
        class_counts = df['label'].value_counts().sort_index()
        class_counts.plot(kind='bar', ax=ax, color=sns.color_palette("viridis", len(class_counts)))
        ax.set_xlabel('Classes')
        ax.set_ylabel('Number of Images')
        ax.set_title(f'{split_name} Set')
        ax.set_xticklabels(class_counts.index, rotation=45, ha='right')
        
        for i, count in enumerate(class_counts):
            ax.text(i, count + 2, str(count), ha='center')
        ax.set_ylim(0, max(class_counts) * 1.2)
    
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

class_distribution_plot(train_dataframe, test_dataframe, validation_dataframe)

# ==============================================================================
# 4. DATA PREPROCESSING (Resize and Save)
# ==============================================================================
def data_preprocess(df, split_name, size, quality=95):
    split_dir = os.path.join(output_dir, split_name)
    os.makedirs(split_dir, exist_ok=True)
    processed_file_paths = []
    for idx, row in tqdm(df.iterrows(), total=len(df), desc=f"Processing {split_name} set"):
        label = row['label']
        label_dir = os.path.join(split_dir, label)
        os.makedirs(label_dir, exist_ok=True)
        img = Image.open(row['file_path']).convert("RGB")
        img = img.resize(size, Image.Resampling.LANCZOS)
        file_name = os.path.basename(row['file_path'])
        save_path = os.path.join(label_dir, file_name)
        img.save(save_path, optimize=True, quality=quality)
        processed_file_paths.append(save_path)
    df['file_path'] = processed_file_paths
    return df

train_dataframe = data_preprocess(train_dataframe, "train", size=image_size)
validation_dataframe = data_preprocess(validation_dataframe, "validation", size=image_size)
test_dataframe = data_preprocess(test_dataframe, "test", size=image_size)

# ==============================================================================
# 5. DATASET CLASS, TRANSFORMS, AND DATALOADERS
# ==============================================================================
class ImageDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe
        self.transform = transform
        unique_labels = sorted(self.dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(unique_labels)}
    def __len__(self):
        return len(self.dataframe)
    def __getitem__(self, index):
        img_path = self.dataframe.iloc[index]['file_path']
        image = Image.open(img_path).convert("RGB")
        label_str = self.dataframe.iloc[index]['label']
        label = self.label_map[label_str]
        if self.transform:
            image = self.transform(image)
        return image, label

def get_dataloaders(size):
    normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    train_transform = transforms.Compose([
        transforms.RandomRotation(degrees=15),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomVerticalFlip(p=0.5),
        transforms.RandomResizedCrop(size, scale=(0.9, 1.0)),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.ToTensor(),
        normalize
    ])
    val_test_transform = transforms.Compose([transforms.Resize(size), transforms.ToTensor(), normalize])
    train_dataset = ImageDataset(train_dataframe, transform=train_transform)
    valid_dataset = ImageDataset(validation_dataframe, transform=val_test_transform)
    test_dataset = ImageDataset(test_dataframe, transform=val_test_transform)
    print("\n--- Label Encoding ---")
    for label, idx in train_dataset.label_map.items():
        print(f"{label} -> {idx}")
    train_dataloader = DataLoader(train_dataset, batch_size=train_batch, shuffle=True, num_workers=2)
    validation_dataloader = DataLoader(valid_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    test_dataloader = DataLoader(test_dataset, batch_size=test_batch, shuffle=False, num_workers=2)
    return train_dataloader, validation_dataloader, test_dataloader, train_dataset.label_map

train_dataloader, validation_dataloader, test_dataloader, label_map = get_dataloaders(size=image_size)
class_names = list(label_map.keys())

# ==============================================================================
# 6. DEEP LEARNING MODEL (Xception)
# ==============================================================================
print("\n--- Initializing Xception Model ---")
# Timm is a great library for PyTorch models. We'll use it to get a pre-trained Xception.
try:
    import timm
    model = timm.create_model('xception', pretrained=True, num_classes=total_class)
except ImportError:
    print("timm not found, installing...")
    !pip install timm --quiet
    import timm
    model = timm.create_model('xception', pretrained=True, num_classes=total_class)

model = model.to(device)

optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=decay)
criterion = nn.CrossEntropyLoss()
summary(model, input_size=(train_batch, 3, 299, 299))

# ==============================================================================
# 7. MODEL TRAINING AND FINE-TUNING
# ==============================================================================
def train_finetune_model(model, criterion, optimizer, train_loader, val_loader, initial_epochs, finetune_epochs, patience, ckpt_path):
    train_loss_history, val_loss_history = [], []
    train_acc_history, val_acc_history = [], []
    best_val_acc = 0.0
    no_improvement_count = 0
    total_start_time = time.time()
    
    # --- STAGE 1: Initial Training (Feature Extraction) ---
    print("\n--- Stage 1: Training the Classifier (Frozen Base) ---")
    # Freeze all layers first
    for param in model.parameters():
        param.requires_grad = False
    # Unfreeze the final classifier layer
    for param in model.fc.parameters():
        param.requires_grad = True

    for epoch in range(initial_epochs):
        model.train()
        running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{initial_epochs} [T]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)
        
        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)

        model.eval()
        running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()
        
        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)
        
        print(f"Epoch {epoch+1}/{initial_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")

        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 1 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
        
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 1 after {patience} epochs with no improvement.")
            break
            
    # --- STAGE 2: Fine-Tuning ---
    print("\n--- Stage 2: Fine-Tuning (Unfrozen Base) ---")
    # Unfreeze all layers
    for param in model.parameters():
        param.requires_grad = True
    
    # Use a much lower learning rate for fine-tuning
    optimizer_ft = optim.Adam(model.parameters(), lr=learning_rate / 10)
    
    no_improvement_count = 0 # Reset counter

    for epoch in range(finetune_epochs):
        model.train()
        running_loss, correct_train, total_train = 0.0, 0, 0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{finetune_epochs} [FT]")
        for inputs, labels in progress_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer_ft.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer_ft.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()
            progress_bar.set_postfix(loss=running_loss/total_train, acc=correct_train/total_train)

        epoch_train_loss = running_loss / total_train
        epoch_train_acc = correct_train / total_train
        train_loss_history.append(epoch_train_loss)
        train_acc_history.append(epoch_train_acc)
        
        model.eval()
        running_val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()

        epoch_val_loss = running_val_loss / total_val
        epoch_val_acc = correct_val / total_val
        val_loss_history.append(epoch_val_loss)
        val_acc_history.append(epoch_val_acc)

        print(f"Epoch {epoch+1}/{finetune_epochs} -> Train Loss: {epoch_train_loss:.3f}, Train Acc: {epoch_train_acc:.3f} | Val Loss: {epoch_val_loss:.3f}, Val Acc: {epoch_val_acc:.3f}")
        
        if epoch_val_acc > best_val_acc:
            best_val_acc = epoch_val_acc
            torch.save(model.state_dict(), os.path.join(ckpt_path, "best_model.pt"))
            print(f"✅ Best model from Stage 2 saved with validation accuracy: {best_val_acc:.3f}")
            no_improvement_count = 0
        else:
            no_improvement_count += 1
            print(f"   (No improvement for {no_improvement_count} epochs)")
            
        if no_improvement_count >= patience:
            print(f"🛑 Early stopping Stage 2 after {patience} epochs with no improvement.")
            break
            
    total_end_time = time.time()
    total_training_time = total_end_time - total_start_time
    print(f"\n--- Training & Fine-Tuning Finished ---")
    print(f"Total Time: {total_training_time:.2f} seconds ({total_training_time/60:.2f} minutes)")

    return train_loss_history, train_acc_history, val_loss_history, val_acc_history

# Train and fine-tune the model
train_loss, train_acc, val_loss, val_acc = train_finetune_model(
    model, criterion, optimizer, train_dataloader, validation_dataloader, 
    initial_epochs, finetune_epochs, patience, checkpoints_path
)

# --- Plot Training History ---
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_loss, label='Train Loss')
plt.plot(val_loss, label='Validation Loss')
plt.axvline(x=initial_epochs-1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Loss over Epochs')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(train_acc, label='Train Accuracy')
plt.plot(val_acc, label='Validation Accuracy')
plt.axvline(x=initial_epochs-1, color='r', linestyle='--', label='Start Fine-Tuning')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Accuracy over Epochs')
plt.legend()
plt.tight_layout()
plt.show()

# ==============================================================================
# 8. EVALUATION
# ==============================================================================
print("\n--- Evaluating Model on Test Set ---")
# Load the best model weights for final evaluation
model.load_state_dict(torch.load(os.path.join(checkpoints_path, "best_model.pt")))

def evaluate_model_full(model, dataloader):
    model.eval()
    y_true, y_pred = [], []
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, predicted = torch.max(outputs, 1)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(predicted.cpu().numpy())
    return np.array(y_true), np.array(y_pred)

y_true, y_pred = evaluate_model_full(model, test_dataloader)

# --- Overall Test Accuracy ---
test_accuracy = accuracy_score(y_true, y_pred)
print(f"Overall Test Accuracy: {test_accuracy:.3f}")

# --- Classification Report ---
print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

# --- Confusion Matrix ---
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()

# ==============================================================================
# 9. EXPLAINABLE AI (Grad-CAM)
# ==============================================================================
class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer
        self.gradients = None
        self.activations = None
        self.hook_handles = []
        self._register_hooks()

    def _register_hooks(self):
        def forward_hook(module, input, output):
            self.activations = output.detach()
        def backward_hook(module, grad_input, grad_output):
            self.gradients = grad_output[0].detach()
        self.hook_handles.extend([
            self.target_layer.register_forward_hook(forward_hook),
            self.target_layer.register_backward_hook(backward_hook)
        ])

    def remove_hooks(self):
        for handle in self.hook_handles:
            handle.remove()

    def generate_cam(self, input_tensor, target_class=None):
        self.model.eval()
        input_tensor.requires_grad_(True)
        output = self.model(input_tensor)
        
        if target_class is None:
            target_class = output.argmax(dim=1).item()
            
        self.model.zero_grad()
        class_score = output[:, target_class]
        class_score.backward()
        
        weights = torch.mean(self.gradients, dim=(2, 3), keepdim=True)
        cam = torch.sum(weights * self.activations, dim=1).squeeze().cpu().numpy()
        cam = np.maximum(cam, 0)
        cam = cv2.resize(cam, (299, 299)) # Resize to Xception size
        cam = (cam - np.min(cam)) / (np.max(cam) - np.min(cam) + 1e-8)
        return cam

def denormalize_image(tensor, mean, std):
    tensor = tensor.clone()
    for t, m, s in zip(tensor, mean, std):
        t.mul_(s).add_(m)
    return tensor

def show_gradcam(model, dataloader, device, target_layer, classes, num_samples=5):
    gradcam = GradCAM(model, target_layer)
    data_iter = iter(dataloader)
    
    plt.figure(figsize=(8, num_samples * 3))
    
    for i in range(num_samples):
        inputs, labels = next(data_iter)
        input_tensor = inputs[0].unsqueeze(0).to(device)
        label = labels[0].item()
        
        output = model(input_tensor)
        pred_class = output.argmax(dim=1).item()

        cam = gradcam.generate_cam(input_tensor, pred_class)
        
        mean_t = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
        std_t = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
        img_denorm = denormalize_image(inputs[0].cpu(), mean_t, std_t)
        img_np = img_denorm.numpy().transpose(1, 2, 0)
        img_np = np.clip(img_np, 0, 1)

        heatmap = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)
        overlay = cv2.addWeighted(cv2.cvtColor(np.uint8(255 * img_np), cv2.COLOR_RGB2BGR), 0.6, heatmap, 0.4, 0)
        
        plt.subplot(num_samples, 2, 2 * i + 1)
        plt.imshow(img_np)
        plt.title(f"Original\nTrue: {classes[label]}")
        plt.axis('off')

        plt.subplot(num_samples, 2, 2 * i + 2)
        plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
        plt.title(f"Grad-CAM\nPredicted: {classes[pred_class]}")
        plt.axis('off')
        
    plt.tight_layout()
    plt.show()
    gradcam.remove_hooks()

print("\n--- Grad-CAM Visualization ---")
# The last convolutional layer in Xception's final block (`conv4`) is a good target.
target_layer = model.conv4
show_gradcam(model, test_dataloader, device, target_layer, class_names, num_samples=5)
```

```python
# =============================================================================
# FULL PIPELINE: INCEPTION V3 vs. MOBILENET V2 (Paired T-Test Comparison)
# =============================================================================

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models, transforms
from torch.utils.data import DataLoader, Dataset
from sklearn.model_selection import train_test_split
from scipy import stats
import pandas as pd
import numpy as np
import os
from PIL import Image
from tqdm.notebook import tqdm
import time

# -----------------------------------------------------------------------------
# 1. CONFIGURATION & SETUP
# -----------------------------------------------------------------------------
# Detect hardware
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Dataset Paths (Exact path from your context)
DATA_DIR = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"
OUTPUT_DIR = "/kaggle/working/" 

# Hyperparameters
NUM_CLASSES = 4
BATCH_SIZE = 32
EPOCHS = 15       # Sufficient for Transfer Learning convergence
LEARNING_RATE = 1e-4
SEED = 42

# Reproducibility
torch.manual_seed(SEED)
np.random.seed(SEED)

# -----------------------------------------------------------------------------
# 2. DATA PREPARATION
# -----------------------------------------------------------------------------
print("\n--- Step 1: Loading and Splitting Data ---")

def load_dataset_metadata(root_dir):
    """Crawls the directory to create a DataFrame of paths and labels."""
    file_paths = []
    labels = []
    
    # Check if directory exists
    if not os.path.exists(root_dir):
        raise FileNotFoundError(f"Directory not found: {root_dir}")
        
    classes = sorted(os.listdir(root_dir))
    print(f"Classes found: {classes}")
    
    for class_name in classes:
        class_dir = os.path.join(root_dir, class_name)
        if os.path.isdir(class_dir):
            for filename in os.listdir(class_dir):
                if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
                    file_paths.append(os.path.join(class_dir, filename))
                    labels.append(class_name)
                    
    return pd.DataFrame({"file_path": file_paths, "label": labels})

# Load Data
df = load_dataset_metadata(DATA_DIR)
print(f"Total images found: {len(df)}")

# Split Data: 70% Train, 15% Validation (unused here but good practice), 15% Test
# We specifically need the TEST set to be isolated for the T-Test.
train_df, temp_df = train_test_split(df, test_size=0.30, stratify=df['label'], random_state=SEED)
val_df, test_df = train_test_split(temp_df, test_size=0.50, stratify=temp_df['label'], random_state=SEED)

print(f"Training Samples: {len(train_df)}")
print(f"Testing Samples:  {len(test_df)} (Used for Paired T-Test)")

# Custom Dataset Class
class BetelLeafDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform
        self.classes = sorted(dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(self.classes)}
        
    def __len__(self):
        return len(self.dataframe)
        
    def __getitem__(self, idx):
        img_path = self.dataframe.iloc[idx]['file_path']
        label_str = self.dataframe.iloc[idx]['label']
        label = self.label_map[label_str]
        
        try:
            image = Image.open(img_path).convert("RGB")
            if self.transform:
                image = self.transform(image)
            return image, label
        except Exception as e:
            print(f"Error loading {img_path}: {e}")
            return torch.zeros(3, 224, 224), label

# -----------------------------------------------------------------------------
# 3. TRANSFORMS & DATALOADERS
# -----------------------------------------------------------------------------
# Note: InceptionV3 requires 299x299 input. MobileNetV2 requires 224x224.
# We create separate transforms for fairness to each architecture.

norm_params = ([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])

# Inception Transforms
train_tf_inception = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])
test_tf_inception = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])

# MobileNet Transforms
train_tf_mobilenet = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])
test_tf_mobilenet = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])

# Create DataLoaders
# Inception Loaders
train_loader_inc = DataLoader(BetelLeafDataset(train_df, train_tf_inception), batch_size=BATCH_SIZE, shuffle=True)
test_loader_inc = DataLoader(BetelLeafDataset(test_df, test_tf_inception), batch_size=BATCH_SIZE, shuffle=False)

# MobileNet Loaders
train_loader_mob = DataLoader(BetelLeafDataset(train_df, train_tf_mobilenet), batch_size=BATCH_SIZE, shuffle=True)
test_loader_mob = DataLoader(BetelLeafDataset(test_df, test_tf_mobilenet), batch_size=BATCH_SIZE, shuffle=False)

# -----------------------------------------------------------------------------
# 4. MODEL DEFINITION & TRAINING
# -----------------------------------------------------------------------------

def build_model(model_name):
    """Initialize model with pre-trained weights and modify head."""
    if model_name == "InceptionV3":
        # Inception V3
        model = models.inception_v3(weights='IMAGENET1K_V1')
        # Handle AuxLogits (Inception specific)
        num_ftrs_aux = model.AuxLogits.fc.in_features
        model.AuxLogits.fc = nn.Linear(num_ftrs_aux, NUM_CLASSES)
        # Handle Primary Head
        num_ftrs = model.fc.in_features
        model.fc = nn.Linear(num_ftrs, NUM_CLASSES)
        
    elif model_name == "MobileNetV2":
        # MobileNet V2
        model = models.mobilenet_v2(weights='IMAGENET1K_V1')
        # MobileNet classifier is a Sequential block, last layer is index 1
        num_ftrs = model.classifier[1].in_features
        model.classifier[1] = nn.Linear(num_ftrs, NUM_CLASSES)
        
    return model.to(device)

def train_model(model, train_loader, model_name):
    """Standard training loop."""
    print(f"\nTraining {model_name} for {EPOCHS} epochs...")
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
    
    model.train()
    
    for epoch in range(EPOCHS):
        running_loss = 0.0
        correct = 0
        total = 0
        
        loop = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS}", leave=False)
        
        for inputs, labels in loop:
            inputs, labels = inputs.to(device), labels.to(device)
            
            optimizer.zero_grad()
            
            # Forward pass
            if model_name == "InceptionV3":
                # Inception returns (outputs, aux_outputs) during training
                outputs, aux_outputs = model(inputs)
                loss1 = criterion(outputs, labels)
                loss2 = criterion(aux_outputs, labels)
                loss = loss1 + 0.4 * loss2
            else:
                # Standard forward for MobileNet
                outputs = model(inputs)
                loss = criterion(outputs, labels)
            
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item()
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
            
            loop.set_postfix(loss=loss.item())
            
        epoch_acc = 100 * correct / total
        print(f"Epoch {epoch+1} - Loss: {running_loss/len(train_loader):.4f} - Acc: {epoch_acc:.2f}%")
        
    return model

# --- EXECUTE TRAINING ---

# 1. Train InceptionV3
model_inc = build_model("InceptionV3")
model_inc = train_model(model_inc, train_loader_inc, "InceptionV3")
# Save weights to ensure we test the exact trained model
torch.save(model_inc.state_dict(), os.path.join(OUTPUT_DIR, "inception_v3_finetuned.pth"))

# 2. Train MobileNetV2
model_mob = build_model("MobileNetV2")
model_mob = train_model(model_mob, train_loader_mob, "MobileNetV2")
torch.save(model_mob.state_dict(), os.path.join(OUTPUT_DIR, "mobilenet_v2_finetuned.pth"))

# -----------------------------------------------------------------------------
# 5. PAIRED T-TEST EVALUATION
# -----------------------------------------------------------------------------
print("\n" + "="*50)
print(" STARTING STATISTICAL COMPARISON (PAIRED T-TEST) ")
print("="*50)

def get_prediction_vector(model, dataloader):
    """
    Runs inference and returns a binary vector where:
    1 = Correct Prediction
    0 = Incorrect Prediction
    This vector is required for the Paired T-Test.
    """
    model.eval()
    binary_results = []
    
    with torch.no_grad():
        for inputs, labels in tqdm(dataloader, desc="Evaluating"):
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            
            # Compare predictions with ground truth
            # Move to CPU and convert to numpy
            batch_results = (preds == labels).cpu().numpy().astype(int)
            binary_results.extend(batch_results)
            
    return np.array(binary_results)

# Load best weights (Optional in this script since variables exist, but good for robustness)
model_inc.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, "inception_v3_finetuned.pth")))
model_mob.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, "mobilenet_v2_finetuned.pth")))

# Get result vectors
print("\nGenerating results for InceptionV3...")
results_inc = get_prediction_vector(model_inc, test_loader_inc)

print("Generating results for MobileNetV2...")
results_mob = get_prediction_vector(model_mob, test_loader_mob)

# Calculate Accuracies
acc_inc = np.mean(results_inc)
acc_mob = np.mean(results_mob)

print(f"\n--- Performance Summary ---")
print(f"InceptionV3 Accuracy:  {acc_inc:.4f} ({acc_inc*100:.2f}%)")
print(f"MobileNetV2 Accuracy:  {acc_mob:.4f} ({acc_mob*100:.2f}%)")

# Perform Paired T-Test
# Null Hypothesis (H0): There is no significant difference between the two models.
t_statistic, p_value = stats.ttest_rel(results_inc, results_mob)

print(f"\n--- Paired T-Test Results ---")
print(f"T-Statistic: {t_statistic:.5f}")
print(f"P-Value:     {p_value:.5f}")

ALPHA = 0.05
print("\n--- Conclusion ---")
if p_value < ALPHA:
    print("The P-Value is less than 0.05.")
    print(">> We REJECT the Null Hypothesis.")
    print(">> There is a STATISTICALLY SIGNIFICANT difference between InceptionV3 and MobileNetV2.")
    if acc_inc > acc_mob:
        print(">> InceptionV3 is the superior model.")
    else:
        print(">> MobileNetV2 is the superior model.")
else:
    print("The P-Value is greater than 0.05.")
    print(">> We FAIL TO REJECT the Null Hypothesis.")
    print(">> There is NO statistically significant difference between the models.")
    print(">> Any difference in accuracy is likely due to chance.")
```

```python
# =============================================================================
# FULL PIPELINE: INCEPTION V3 vs. DENSENET 201 (Robust & Optimized)
# =============================================================================

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models, transforms
from torch.utils.data import DataLoader, Dataset
from sklearn.model_selection import train_test_split
from scipy import stats
import pandas as pd
import numpy as np
import os
from PIL import Image
import time

# -----------------------------------------------------------------------------
# 1. CONFIGURATION & SETUP
# -----------------------------------------------------------------------------
# Detect hardware
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

if device.type == 'cpu':
    print("WARNING: You are training DenseNet201 on CPU. This will be very slow.")
    print("RECOMMENDATION: Enable GPU Accelerator in Notebook Settings.")

# Dataset Paths
DATA_DIR = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"
OUTPUT_DIR = "/kaggle/working/" 

# Hyperparameters
NUM_CLASSES = 4
BATCH_SIZE = 32
EPOCHS = 15       
LEARNING_RATE = 1e-4
SEED = 42

# Reproducibility
torch.manual_seed(SEED)
np.random.seed(SEED)
if torch.cuda.is_available():
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

# -----------------------------------------------------------------------------
# 2. DATA PREPARATION
# -----------------------------------------------------------------------------
print("\n--- Step 1: Loading and Splitting Data ---")

def load_dataset_metadata(root_dir):
    """Crawls the directory to create a DataFrame of paths and labels."""
    file_paths = []
    labels = []
    
    if not os.path.exists(root_dir):
        raise FileNotFoundError(f"Directory not found: {root_dir}")
        
    classes = sorted(os.listdir(root_dir))
    print(f"Classes found: {classes}")
    
    for class_name in classes:
        class_dir = os.path.join(root_dir, class_name)
        if os.path.isdir(class_dir):
            for filename in os.listdir(class_dir):
                if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
                    file_paths.append(os.path.join(class_dir, filename))
                    labels.append(class_name)
                    
    return pd.DataFrame({"file_path": file_paths, "label": labels})

# Load Data
df = load_dataset_metadata(DATA_DIR)
print(f"Total images found: {len(df)}")

# Split Data: 70% Train, 30% Temp (15% Val, 15% Test)
train_df, temp_df = train_test_split(df, test_size=0.30, stratify=df['label'], random_state=SEED)
val_df, test_df = train_test_split(temp_df, test_size=0.50, stratify=temp_df['label'], random_state=SEED)

print(f"Training Samples: {len(train_df)}")
print(f"Testing Samples:  {len(test_df)}")

# Custom Dataset Class
class BetelLeafDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform
        self.classes = sorted(dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(self.classes)}
        
    def __len__(self):
        return len(self.dataframe)
        
    def __getitem__(self, idx):
        img_path = self.dataframe.iloc[idx]['file_path']
        label_str = self.dataframe.iloc[idx]['label']
        label = self.label_map[label_str]
        
        try:
            image = Image.open(img_path).convert("RGB")
            if self.transform:
                image = self.transform(image)
            return image, label
        except Exception as e:
            print(f"Error loading {img_path}: {e}")
            return torch.zeros(3, 224, 224), label

# -----------------------------------------------------------------------------
# 3. TRANSFORMS & DATALOADERS
# -----------------------------------------------------------------------------
norm_params = ([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])

# Inception V3 Transforms (299x299)
train_tf_inc = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])
test_tf_inc = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])

# DenseNet201 Transforms (224x224)
train_tf_dense = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])
test_tf_dense = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])

# Dataloaders - Optimized with num_workers
num_workers = 2 if device.type == 'cpu' else 4
pin_memory = True if device.type == 'cuda' else False

train_loader_inc = DataLoader(BetelLeafDataset(train_df, train_tf_inc), 
                              batch_size=BATCH_SIZE, shuffle=True, num_workers=num_workers, pin_memory=pin_memory)
test_loader_inc = DataLoader(BetelLeafDataset(test_df, test_tf_inc), 
                             batch_size=BATCH_SIZE, shuffle=False, num_workers=num_workers, pin_memory=pin_memory)

train_loader_dense = DataLoader(BetelLeafDataset(train_df, train_tf_dense), 
                                batch_size=BATCH_SIZE, shuffle=True, num_workers=num_workers, pin_memory=pin_memory)
test_loader_dense = DataLoader(BetelLeafDataset(test_df, test_tf_dense), 
                               batch_size=BATCH_SIZE, shuffle=False, num_workers=num_workers, pin_memory=pin_memory)

# -----------------------------------------------------------------------------
# 4. MODEL DEFINITION & TRAINING
# -----------------------------------------------------------------------------

def build_model(model_name):
    print(f"Building {model_name}...")
    if model_name == "InceptionV3":
        model = models.inception_v3(weights='IMAGENET1K_V1')
        num_ftrs_aux = model.AuxLogits.fc.in_features
        model.AuxLogits.fc = nn.Linear(num_ftrs_aux, NUM_CLASSES)
        num_ftrs = model.fc.in_features
        model.fc = nn.Linear(num_ftrs, NUM_CLASSES)
    elif model_name == "DenseNet201":
        model = models.densenet201(weights='IMAGENET1K_V1')
        num_ftrs = model.classifier.in_features
        model.classifier = nn.Linear(num_ftrs, NUM_CLASSES)
    return model.to(device)

def train_model(model, train_loader, model_name):
    print(f"\n--- Training {model_name} ---")
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
    
    for epoch in range(EPOCHS):
        model.train()
        running_loss = 0.0
        correct = 0
        total = 0
        start_time = time.time()
        
        # Using a simple loop instead of tqdm per batch to avoid notebook hanging/freezing
        for i, (inputs, labels) in enumerate(train_loader):
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            
            if model_name == "InceptionV3":
                outputs, aux_outputs = model(inputs)
                loss1 = criterion(outputs, labels)
                loss2 = criterion(aux_outputs, labels)
                loss = loss1 + 0.4 * loss2
            else:
                outputs = model(inputs)
                loss = criterion(outputs, labels)
            
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item()
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
            
        epoch_time = time.time() - start_time
        epoch_acc = 100 * correct / total
        print(f"Epoch {epoch+1}/{EPOCHS} | Time: {epoch_time:.0f}s | Loss: {running_loss/len(train_loader):.4f} | Acc: {epoch_acc:.2f}%")
        
    return model

# --- EXECUTE TRAINING ---

# 1. Train InceptionV3
model_inc = build_model("InceptionV3")
model_inc = train_model(model_inc, train_loader_inc, "InceptionV3")
torch.save(model_inc.state_dict(), os.path.join(OUTPUT_DIR, "inception_v3_final.pth"))

# 2. Train DenseNet201
model_dense = build_model("DenseNet201")
model_dense = train_model(model_dense, train_loader_dense, "DenseNet201")
torch.save(model_dense.state_dict(), os.path.join(OUTPUT_DIR, "densenet201_final.pth"))

# -----------------------------------------------------------------------------
# 5. PAIRED T-TEST EVALUATION
# -----------------------------------------------------------------------------
print("\n" + "="*50)
print(" STARTING STATISTICAL COMPARISON (PAIRED T-TEST) ")
print("="*50)

def get_prediction_vector(model, dataloader):
    model.eval()
    binary_results = []
    
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            results = (preds == labels).cpu().numpy().astype(int)
            binary_results.extend(results)
            
    return np.array(binary_results)

# Reload Best Weights
model_inc.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, "inception_v3_final.pth")))
model_dense.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, "densenet201_final.pth")))

print("Evaluating InceptionV3...")
results_inc = get_prediction_vector(model_inc, test_loader_inc)

print("Evaluating DenseNet201...")
results_dense = get_prediction_vector(model_dense, test_loader_dense)

# Accuracy Calculation
acc_inc = np.mean(results_inc)
acc_dense = np.mean(results_dense)

print(f"\nInceptionV3 Accuracy:  {acc_inc:.4f} ({acc_inc*100:.2f}%)")
print(f"DenseNet201 Accuracy:  {acc_dense:.4f} ({acc_dense*100:.2f}%)")

# Paired T-Test
t_stat, p_val = stats.ttest_rel(results_inc, results_dense)

print(f"\n--- Results ---")
print(f"T-Statistic: {t_stat:.5f}")
print(f"P-Value:     {p_val:.5f}")

if p_val < 0.05:
    print("Conclusion: SIGNIFICANT DIFFERENCE found between models.")
else:
    print("Conclusion: NO SIGNIFICANT DIFFERENCE found.")
```

```python
# =============================================================================
# FULL PIPELINE: INCEPTION V3 vs. EFFICIENTNET B0 (Paired T-Test Comparison)
# =============================================================================

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models, transforms
from torch.utils.data import DataLoader, Dataset
from sklearn.model_selection import train_test_split
from scipy import stats
import pandas as pd
import numpy as np
import os
from PIL import Image
from tqdm.notebook import tqdm

# -----------------------------------------------------------------------------
# 1. CONFIGURATION & SETUP
# -----------------------------------------------------------------------------
# Detect hardware
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Dataset Paths
DATA_DIR = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"
OUTPUT_DIR = "/kaggle/working/" 

# Hyperparameters
NUM_CLASSES = 4
BATCH_SIZE = 32
EPOCHS = 15       
LEARNING_RATE = 1e-4
SEED = 42

# Reproducibility
torch.manual_seed(SEED)
np.random.seed(SEED)

# -----------------------------------------------------------------------------
# 2. DATA PREPARATION
# -----------------------------------------------------------------------------
print("\n--- Step 1: Loading and Splitting Data ---")

def load_dataset_metadata(root_dir):
    """Crawls the directory to create a DataFrame of paths and labels."""
    file_paths = []
    labels = []
    
    if not os.path.exists(root_dir):
        raise FileNotFoundError(f"Directory not found: {root_dir}")
        
    classes = sorted(os.listdir(root_dir))
    print(f"Classes found: {classes}")
    
    for class_name in classes:
        class_dir = os.path.join(root_dir, class_name)
        if os.path.isdir(class_dir):
            for filename in os.listdir(class_dir):
                if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
                    file_paths.append(os.path.join(class_dir, filename))
                    labels.append(class_name)
                    
    return pd.DataFrame({"file_path": file_paths, "label": labels})

# Load Data
df = load_dataset_metadata(DATA_DIR)
print(f"Total images found: {len(df)}")

# Split Data: 70% Train, 15% Val (unused here), 15% Test
# The TEST set is isolated strictly for the T-Test
train_df, temp_df = train_test_split(df, test_size=0.30, stratify=df['label'], random_state=SEED)
val_df, test_df = train_test_split(temp_df, test_size=0.50, stratify=temp_df['label'], random_state=SEED)

print(f"Training Samples: {len(train_df)}")
print(f"Testing Samples:  {len(test_df)} (Used for Paired T-Test)")

# Custom Dataset Class
class BetelLeafDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform
        self.classes = sorted(dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(self.classes)}
        
    def __len__(self):
        return len(self.dataframe)
        
    def __getitem__(self, idx):
        img_path = self.dataframe.iloc[idx]['file_path']
        label_str = self.dataframe.iloc[idx]['label']
        label = self.label_map[label_str]
        
        try:
            image = Image.open(img_path).convert("RGB")
            if self.transform:
                image = self.transform(image)
            return image, label
        except Exception as e:
            print(f"Error loading {img_path}: {e}")
            return torch.zeros(3, 224, 224), label

# -----------------------------------------------------------------------------
# 3. TRANSFORMS & DATALOADERS
# -----------------------------------------------------------------------------
# InceptionV3 requires 299x299.
# EfficientNetB0 standard resolution is 224x224.

norm_params = ([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])

# --- Inception V3 Transforms (299x299) ---
train_tf_inc = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])
test_tf_inc = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])

# --- EfficientNet B0 Transforms (224x224) ---
train_tf_eff = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])
test_tf_eff = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])

# Create DataLoaders
# Inception Loaders
train_loader_inc = DataLoader(BetelLeafDataset(train_df, train_tf_inc), batch_size=BATCH_SIZE, shuffle=True)
test_loader_inc = DataLoader(BetelLeafDataset(test_df, test_tf_inc), batch_size=BATCH_SIZE, shuffle=False)

# EfficientNet Loaders
train_loader_eff = DataLoader(BetelLeafDataset(train_df, train_tf_eff), batch_size=BATCH_SIZE, shuffle=True)
test_loader_eff = DataLoader(BetelLeafDataset(test_df, test_tf_eff), batch_size=BATCH_SIZE, shuffle=False)

# -----------------------------------------------------------------------------
# 4. MODEL DEFINITION & TRAINING
# -----------------------------------------------------------------------------

def build_model(model_name):
    """Initialize model with pre-trained weights and modify classifier head."""
    print(f"Building {model_name}...")
    
    if model_name == "InceptionV3":
        # Inception V3 (requires aux logits handling)
        model = models.inception_v3(weights='IMAGENET1K_V1')
        
        # Modify Aux Head
        num_ftrs_aux = model.AuxLogits.fc.in_features
        model.AuxLogits.fc = nn.Linear(num_ftrs_aux, NUM_CLASSES)
        
        # Modify Primary Head
        num_ftrs = model.fc.in_features
        model.fc = nn.Linear(num_ftrs, NUM_CLASSES)
        
    elif model_name == "EfficientNetB0":
        # EfficientNet B0
        model = models.efficientnet_b0(weights='IMAGENET1K_V1')
        
        # EfficientNet classifier is a Sequential block: (Dropout, Linear)
        # We access the last layer (Linear) usually at index 1
        num_ftrs = model.classifier[1].in_features
        model.classifier[1] = nn.Linear(num_ftrs, NUM_CLASSES)
        
    return model.to(device)

def train_model(model, train_loader, model_name):
    """Training loop that handles Inception's specific output format."""
    print(f"\n--- Training {model_name} ---")
    
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
    model.train()
    
    for epoch in range(EPOCHS):
        running_loss = 0.0
        correct = 0
        total = 0
        
        loop = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS}", leave=False)
        
        for inputs, labels in loop:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            
            # Forward pass
            if model_name == "InceptionV3":
                # Inception returns (outputs, aux_outputs) during training
                outputs, aux_outputs = model(inputs)
                loss1 = criterion(outputs, labels)
                loss2 = criterion(aux_outputs, labels)
                loss = loss1 + 0.4 * loss2
            else:
                # EfficientNet returns single output
                outputs = model(inputs)
                loss = criterion(outputs, labels)
            
            loss.backward()
            optimizer.step()
            
            # Stats
            running_loss += loss.item()
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
            
            loop.set_postfix(loss=loss.item())
            
        epoch_acc = 100 * correct / total
        print(f"Epoch {epoch+1} - Avg Loss: {running_loss/len(train_loader):.4f} - Acc: {epoch_acc:.2f}%")
        
    return model

# --- EXECUTE TRAINING ---

# 1. Train InceptionV3
model_inc = build_model("InceptionV3")
model_inc = train_model(model_inc, train_loader_inc, "InceptionV3")
torch.save(model_inc.state_dict(), os.path.join(OUTPUT_DIR, "inception_v3_final.pth"))

# 2. Train EfficientNetB0
model_eff = build_model("EfficientNetB0")
model_eff = train_model(model_eff, train_loader_eff, "EfficientNetB0")
torch.save(model_eff.state_dict(), os.path.join(OUTPUT_DIR, "efficientnet_b0_final.pth"))

# -----------------------------------------------------------------------------
# 5. PAIRED T-TEST EVALUATION
# -----------------------------------------------------------------------------
print("\n" + "="*50)
print(" STARTING STATISTICAL COMPARISON (PAIRED T-TEST) ")
print("="*50)

def get_prediction_vector(model, dataloader):
    """
    Returns a binary numpy array:
    1 = Correct Prediction
    0 = Incorrect Prediction
    """
    model.eval()
    binary_results = []
    
    with torch.no_grad():
        for inputs, labels in tqdm(dataloader, desc="Evaluating"):
            inputs, labels = inputs.to(device), labels.to(device)
            
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            
            # Compare preds with actual labels
            results = (preds == labels).cpu().numpy().astype(int)
            binary_results.extend(results)
            
    return np.array(binary_results)

# Ensure models are loaded with best/final weights
model_inc.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, "inception_v3_final.pth")))
model_eff.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, "efficientnet_b0_final.pth")))

# Get Binary Result Vectors
print("\nEvaluating InceptionV3 on Test Set...")
results_inc = get_prediction_vector(model_inc, test_loader_inc)

print("Evaluating EfficientNetB0 on Test Set...")
results_eff = get_prediction_vector(model_eff, test_loader_eff)

# Calculate Mean Accuracy
acc_inc = np.mean(results_inc)
acc_eff = np.mean(results_eff)

print(f"\n--- Model Performance ---")
print(f"InceptionV3 Accuracy:    {acc_inc:.4f} ({acc_inc*100:.2f}%)")
print(f"EfficientNetB0 Accuracy: {acc_eff:.4f} ({acc_eff*100:.2f}%)")

# Perform Paired T-Test
# Checks if the difference between the two result vectors is statistically significant
t_stat, p_val = stats.ttest_rel(results_inc, results_eff)

print(f"\n--- Paired T-Test Results ---")
print(f"T-Statistic: {t_stat:.5f}")
print(f"P-Value:     {p_val:.5f}")

ALPHA = 0.05
print("\n--- Final Conclusion ---")
if p_val < ALPHA:
    print("Result: SIGNIFICANT DIFFERENCE")
    print(f"Since P-Value ({p_val:.5f}) < {ALPHA}, we reject the null hypothesis.")
    if acc_inc > acc_eff:
        print(">> InceptionV3 significantly outperforms EfficientNetB0.")
    else:
        print(">> EfficientNetB0 significantly outperforms InceptionV3.")
else:
    print("Result: NO SIGNIFICANT DIFFERENCE")
    print(f"Since P-Value ({p_val:.5f}) > {ALPHA}, we fail to reject the null hypothesis.")
    print(">> The performance difference is likely due to random chance.")
```

```python
# =============================================================================
# FULL PIPELINE: INCEPTION V3 vs. ALEXNET (Paired T-Test Comparison)
# =============================================================================

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models, transforms
from torch.utils.data import DataLoader, Dataset
from sklearn.model_selection import train_test_split
from scipy import stats
import pandas as pd
import numpy as np
import os
from PIL import Image
import time

# -----------------------------------------------------------------------------
# 1. CONFIGURATION & SETUP
# -----------------------------------------------------------------------------
# Detect hardware
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

if device.type == 'cpu':
    print("WARNING: Training is running on CPU. This will be slow.")
    print("RECOMMENDATION: Enable GPU Accelerator in Notebook Settings.")

# Dataset Paths
DATA_DIR = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"
OUTPUT_DIR = "/kaggle/working/" 

# Hyperparameters
NUM_CLASSES = 4
BATCH_SIZE = 32
EPOCHS = 15       
LEARNING_RATE = 1e-4
SEED = 42

# Reproducibility
torch.manual_seed(SEED)
np.random.seed(SEED)
if torch.cuda.is_available():
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

# -----------------------------------------------------------------------------
# 2. DATA PREPARATION
# -----------------------------------------------------------------------------
print("\n--- Step 1: Loading and Splitting Data ---")

def load_dataset_metadata(root_dir):
    """Crawls the directory to create a DataFrame of paths and labels."""
    file_paths = []
    labels = []
    
    if not os.path.exists(root_dir):
        raise FileNotFoundError(f"Directory not found: {root_dir}")
        
    classes = sorted(os.listdir(root_dir))
    print(f"Classes found: {classes}")
    
    for class_name in classes:
        class_dir = os.path.join(root_dir, class_name)
        if os.path.isdir(class_dir):
            for filename in os.listdir(class_dir):
                if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
                    file_paths.append(os.path.join(class_dir, filename))
                    labels.append(class_name)
                    
    return pd.DataFrame({"file_path": file_paths, "label": labels})

# Load Data
df = load_dataset_metadata(DATA_DIR)
print(f"Total images found: {len(df)}")

# Split Data: 70% Train, 30% Temp (15% Val, 15% Test)
train_df, temp_df = train_test_split(df, test_size=0.30, stratify=df['label'], random_state=SEED)
val_df, test_df = train_test_split(temp_df, test_size=0.50, stratify=temp_df['label'], random_state=SEED)

print(f"Training Samples: {len(train_df)}")
print(f"Testing Samples:  {len(test_df)} (Used for Paired T-Test)")

# Custom Dataset Class
class BetelLeafDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform
        self.classes = sorted(dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(self.classes)}
        
    def __len__(self):
        return len(self.dataframe)
        
    def __getitem__(self, idx):
        img_path = self.dataframe.iloc[idx]['file_path']
        label_str = self.dataframe.iloc[idx]['label']
        label = self.label_map[label_str]
        
        try:
            image = Image.open(img_path).convert("RGB")
            if self.transform:
                image = self.transform(image)
            return image, label
        except Exception as e:
            print(f"Error loading {img_path}: {e}")
            return torch.zeros(3, 224, 224), label

# -----------------------------------------------------------------------------
# 3. TRANSFORMS & DATALOADERS
# -----------------------------------------------------------------------------
norm_params = ([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])

# Inception V3 Transforms (299x299)
train_tf_inc = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])
test_tf_inc = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])

# AlexNet Transforms (224x224)
train_tf_alex = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])
test_tf_alex = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])

# Dataloaders
num_workers = 2 if device.type == 'cpu' else 4
pin_memory = True if device.type == 'cuda' else False

train_loader_inc = DataLoader(BetelLeafDataset(train_df, train_tf_inc), 
                              batch_size=BATCH_SIZE, shuffle=True, num_workers=num_workers, pin_memory=pin_memory)
test_loader_inc = DataLoader(BetelLeafDataset(test_df, test_tf_inc), 
                             batch_size=BATCH_SIZE, shuffle=False, num_workers=num_workers, pin_memory=pin_memory)

train_loader_alex = DataLoader(BetelLeafDataset(train_df, train_tf_alex), 
                               batch_size=BATCH_SIZE, shuffle=True, num_workers=num_workers, pin_memory=pin_memory)
test_loader_alex = DataLoader(BetelLeafDataset(test_df, test_tf_alex), 
                              batch_size=BATCH_SIZE, shuffle=False, num_workers=num_workers, pin_memory=pin_memory)

# -----------------------------------------------------------------------------
# 4. MODEL DEFINITION & TRAINING
# -----------------------------------------------------------------------------

def build_model(model_name):
    print(f"Building {model_name}...")
    
    if model_name == "InceptionV3":
        # Inception V3 (GoogleNet variant)
        model = models.inception_v3(weights='IMAGENET1K_V1')
        # Aux Head
        num_ftrs_aux = model.AuxLogits.fc.in_features
        model.AuxLogits.fc = nn.Linear(num_ftrs_aux, NUM_CLASSES)
        # Main Head
        num_ftrs = model.fc.in_features
        model.fc = nn.Linear(num_ftrs, NUM_CLASSES)
        
    elif model_name == "AlexNet":
        # AlexNet (Classic architecture)
        model = models.alexnet(weights='IMAGENET1K_V1')
        # Classifier is a Sequential block. The last layer is at index 6.
        # Structure: (0): Dropout, (1): Linear, (2): ReLU, (3): Dropout, (4): Linear, (5): ReLU, (6): Linear
        num_ftrs = model.classifier[6].in_features
        model.classifier[6] = nn.Linear(num_ftrs, NUM_CLASSES)
        
    return model.to(device)

def train_model(model, train_loader, model_name):
    print(f"\n--- Training {model_name} ---")
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
    
    for epoch in range(EPOCHS):
        model.train()
        running_loss = 0.0
        correct = 0
        total = 0
        start_time = time.time()
        
        for i, (inputs, labels) in enumerate(train_loader):
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            
            # Forward pass
            if model_name == "InceptionV3":
                # Inception returns tuple (outputs, aux_outputs) in train mode
                outputs, aux_outputs = model(inputs)
                loss1 = criterion(outputs, labels)
                loss2 = criterion(aux_outputs, labels)
                loss = loss1 + 0.4 * loss2
            else:
                # AlexNet returns single output
                outputs = model(inputs)
                loss = criterion(outputs, labels)
            
            loss.backward()
            optimizer.step()
            
            # Statistics
            running_loss += loss.item()
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
            
        epoch_time = time.time() - start_time
        epoch_acc = 100 * correct / total
        print(f"Epoch {epoch+1}/{EPOCHS} | Time: {epoch_time:.0f}s | Loss: {running_loss/len(train_loader):.4f} | Acc: {epoch_acc:.2f}%")
        
    return model

# --- EXECUTE TRAINING ---

# 1. Train InceptionV3
model_inc = build_model("InceptionV3")
model_inc = train_model(model_inc, train_loader_inc, "InceptionV3")
torch.save(model_inc.state_dict(), os.path.join(OUTPUT_DIR, "inception_v3_final.pth"))

# 2. Train AlexNet
model_alex = build_model("AlexNet")
model_alex = train_model(model_alex, train_loader_alex, "AlexNet")
torch.save(model_alex.state_dict(), os.path.join(OUTPUT_DIR, "alexnet_final.pth"))

# -----------------------------------------------------------------------------
# 5. PAIRED T-TEST EVALUATION
# -----------------------------------------------------------------------------
print("\n" + "="*50)
print(" STARTING STATISTICAL COMPARISON (PAIRED T-TEST) ")
print("="*50)

def get_prediction_vector(model, dataloader):
    """
    Returns a binary numpy array (1=Correct, 0=Incorrect)
    """
    model.eval()
    binary_results = []
    
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            results = (preds == labels).cpu().numpy().astype(int)
            binary_results.extend(results)
            
    return np.array(binary_results)

# Reload Best Weights
model_inc.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, "inception_v3_final.pth")))
model_alex.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, "alexnet_final.pth")))

print("Evaluating InceptionV3...")
results_inc = get_prediction_vector(model_inc, test_loader_inc)

print("Evaluating AlexNet...")
results_alex = get_prediction_vector(model_alex, test_loader_alex)

# Accuracy Calculation
acc_inc = np.mean(results_inc)
acc_alex = np.mean(results_alex)

print(f"\n--- Model Performance ---")
print(f"InceptionV3 Accuracy: {acc_inc:.4f} ({acc_inc*100:.2f}%)")
print(f"AlexNet Accuracy:     {acc_alex:.4f} ({acc_alex*100:.2f}%)")

# Paired T-Test
# Checks if the mean difference between the two sets of observations is zero.
t_stat, p_val = stats.ttest_rel(results_inc, results_alex)

print(f"\n--- Paired T-Test Results ---")
print(f"T-Statistic: {t_stat:.5f}")
print(f"P-Value:     {p_val:.5f}")

ALPHA = 0.05
print("\n--- Conclusion ---")
if p_val < ALPHA:
    print("Result: SIGNIFICANT DIFFERENCE")
    print(f"Since P-Value ({p_val:.5f}) < {ALPHA}, we REJECT the Null Hypothesis.")
    if acc_inc > acc_alex:
        print(">> InceptionV3 significantly outperforms AlexNet.")
    else:
        print(">> AlexNet significantly outperforms InceptionV3.")
else:
    print("Result: NO SIGNIFICANT DIFFERENCE")
    print(f"Since P-Value ({p_val:.5f}) > {ALPHA}, we FAIL TO REJECT the Null Hypothesis.")
    print(">> The models perform similarly on this test set.")
```

```python
# =============================================================================
# FULL PIPELINE: INCEPTION V3 vs. VGG16 (Paired T-Test Comparison)
# =============================================================================

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models, transforms
from torch.utils.data import DataLoader, Dataset
from sklearn.model_selection import train_test_split
from scipy import stats
import pandas as pd
import numpy as np
import os
from PIL import Image
import time

# -----------------------------------------------------------------------------
# 1. CONFIGURATION & SETUP
# -----------------------------------------------------------------------------
# Detect hardware
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

if device.type == 'cpu':
    print("WARNING: Training is running on CPU. This will be very slow.")
    print("RECOMMENDATION: Enable GPU Accelerator in Notebook Settings.")

# Dataset Paths
DATA_DIR = "/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images"
OUTPUT_DIR = "/kaggle/working/" 

# Hyperparameters
NUM_CLASSES = 4
BATCH_SIZE = 32
EPOCHS = 15       
LEARNING_RATE = 1e-4
SEED = 42

# Reproducibility
torch.manual_seed(SEED)
np.random.seed(SEED)
if torch.cuda.is_available():
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

# -----------------------------------------------------------------------------
# 2. DATA PREPARATION
# -----------------------------------------------------------------------------
print("\n--- Step 1: Loading and Splitting Data ---")

def load_dataset_metadata(root_dir):
    """Crawls the directory to create a DataFrame of paths and labels."""
    file_paths = []
    labels = []
    
    if not os.path.exists(root_dir):
        raise FileNotFoundError(f"Directory not found: {root_dir}")
        
    classes = sorted(os.listdir(root_dir))
    print(f"Classes found: {classes}")
    
    for class_name in classes:
        class_dir = os.path.join(root_dir, class_name)
        if os.path.isdir(class_dir):
            for filename in os.listdir(class_dir):
                if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
                    file_paths.append(os.path.join(class_dir, filename))
                    labels.append(class_name)
                    
    return pd.DataFrame({"file_path": file_paths, "label": labels})

# Load Data
df = load_dataset_metadata(DATA_DIR)
print(f"Total images found: {len(df)}")

# Split Data: 70% Train, 30% Temp (15% Val, 15% Test)
# We strictly isolate the Test set for the final T-Test
train_df, temp_df = train_test_split(df, test_size=0.30, stratify=df['label'], random_state=SEED)
val_df, test_df = train_test_split(temp_df, test_size=0.50, stratify=temp_df['label'], random_state=SEED)

print(f"Training Samples: {len(train_df)}")
print(f"Testing Samples:  {len(test_df)} (Used for Paired T-Test)")

# Custom Dataset Class
class BetelLeafDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform
        self.classes = sorted(dataframe['label'].unique())
        self.label_map = {label: idx for idx, label in enumerate(self.classes)}
        
    def __len__(self):
        return len(self.dataframe)
        
    def __getitem__(self, idx):
        img_path = self.dataframe.iloc[idx]['file_path']
        label_str = self.dataframe.iloc[idx]['label']
        label = self.label_map[label_str]
        
        try:
            image = Image.open(img_path).convert("RGB")
            if self.transform:
                image = self.transform(image)
            return image, label
        except Exception as e:
            print(f"Error loading {img_path}: {e}")
            # Return dummy image in case of corruption
            return torch.zeros(3, 224, 224), label

# -----------------------------------------------------------------------------
# 3. TRANSFORMS & DATALOADERS
# -----------------------------------------------------------------------------
norm_params = ([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])

# --- Inception V3 Transforms (299x299) ---
train_tf_inc = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])
test_tf_inc = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])

# --- VGG16 Transforms (224x224) ---
train_tf_vgg = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])
test_tf_vgg = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(*norm_params)
])

# Dataloaders
num_workers = 2 if device.type == 'cpu' else 4
pin_memory = True if device.type == 'cuda' else False

# Inception Loaders
train_loader_inc = DataLoader(BetelLeafDataset(train_df, train_tf_inc), 
                              batch_size=BATCH_SIZE, shuffle=True, num_workers=num_workers, pin_memory=pin_memory)
test_loader_inc = DataLoader(BetelLeafDataset(test_df, test_tf_inc), 
                             batch_size=BATCH_SIZE, shuffle=False, num_workers=num_workers, pin_memory=pin_memory)

# VGG16 Loaders
train_loader_vgg = DataLoader(BetelLeafDataset(train_df, train_tf_vgg), 
                              batch_size=BATCH_SIZE, shuffle=True, num_workers=num_workers, pin_memory=pin_memory)
test_loader_vgg = DataLoader(BetelLeafDataset(test_df, test_tf_vgg), 
                             batch_size=BATCH_SIZE, shuffle=False, num_workers=num_workers, pin_memory=pin_memory)

# -----------------------------------------------------------------------------
# 4. MODEL DEFINITION & TRAINING
# -----------------------------------------------------------------------------

def build_model(model_name):
    print(f"Building {model_name}...")
    
    if model_name == "InceptionV3":
        # Inception V3
        model = models.inception_v3(weights='IMAGENET1K_V1')
        
        # Modify Aux Head
        num_ftrs_aux = model.AuxLogits.fc.in_features
        model.AuxLogits.fc = nn.Linear(num_ftrs_aux, NUM_CLASSES)
        
        # Modify Main Head
        num_ftrs = model.fc.in_features
        model.fc = nn.Linear(num_ftrs, NUM_CLASSES)
        
    elif model_name == "VGG16":
        # VGG16
        model = models.vgg16(weights='IMAGENET1K_V1')
        
        # VGG Classifier is a Sequential block:
        # (0): Linear, (1): ReLU, (2): Dropout, ... (6): Linear(4096, 1000)
        num_ftrs = model.classifier[6].in_features
        model.classifier[6] = nn.Linear(num_ftrs, NUM_CLASSES)
        
    return model.to(device)

def train_model(model, train_loader, model_name):
    print(f"\n--- Training {model_name} ---")
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
    
    for epoch in range(EPOCHS):
        model.train()
        running_loss = 0.0
        correct = 0
        total = 0
        start_time = time.time()
        
        for i, (inputs, labels) in enumerate(train_loader):
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            
            # Forward pass
            if model_name == "InceptionV3":
                # Inception returns (logits, aux_logits) in train mode
                outputs, aux_outputs = model(inputs)
                loss1 = criterion(outputs, labels)
                loss2 = criterion(aux_outputs, labels)
                loss = loss1 + 0.4 * loss2
            else:
                # VGG16 returns single output
                outputs = model(inputs)
                loss = criterion(outputs, labels)
            
            loss.backward()
            optimizer.step()
            
            # Statistics
            running_loss += loss.item()
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
            
        epoch_time = time.time() - start_time
        epoch_acc = 100 * correct / total
        print(f"Epoch {epoch+1}/{EPOCHS} | Time: {epoch_time:.0f}s | Loss: {running_loss/len(train_loader):.4f} | Acc: {epoch_acc:.2f}%")
        
    return model

# --- EXECUTE TRAINING ---

# 1. Train InceptionV3
model_inc = build_model("InceptionV3")
model_inc = train_model(model_inc, train_loader_inc, "InceptionV3")
torch.save(model_inc.state_dict(), os.path.join(OUTPUT_DIR, "inception_v3_final.pth"))

# 2. Train VGG16
model_vgg = build_model("VGG16")
model_vgg = train_model(model_vgg, train_loader_vgg, "VGG16")
torch.save(model_vgg.state_dict(), os.path.join(OUTPUT_DIR, "vgg16_final.pth"))

# -----------------------------------------------------------------------------
# 5. PAIRED T-TEST EVALUATION
# -----------------------------------------------------------------------------
print("\n" + "="*50)
print(" STARTING STATISTICAL COMPARISON (PAIRED T-TEST) ")
print("="*50)

def get_prediction_vector(model, dataloader):
    """
    Returns a binary numpy array (1=Correct, 0=Incorrect)
    """
    model.eval()
    binary_results = []
    
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            results = (preds == labels).cpu().numpy().astype(int)
            binary_results.extend(results)
            
    return np.array(binary_results)

# Reload Best Weights
model_inc.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, "inception_v3_final.pth")))
model_vgg.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, "vgg16_final.pth")))

print("Evaluating InceptionV3...")
results_inc = get_prediction_vector(model_inc, test_loader_inc)

print("Evaluating VGG16...")
results_vgg = get_prediction_vector(model_vgg, test_loader_vgg)

# Accuracy Calculation
acc_inc = np.mean(results_inc)
acc_vgg = np.mean(results_vgg)

print(f"\n--- Model Performance ---")
print(f"InceptionV3 Accuracy: {acc_inc:.4f} ({acc_inc*100:.2f}%)")
print(f"VGG16 Accuracy:       {acc_vgg:.4f} ({acc_vgg*100:.2f}%)")

# Paired T-Test
# Checks if the mean difference between the two sets of observations is zero.
t_stat, p_val = stats.ttest_rel(results_inc, results_vgg)

print(f"\n--- Paired T-Test Results ---")
print(f"T-Statistic: {t_stat:.5f}")
print(f"P-Value:     {p_val:.5f}")

ALPHA = 0.05
print("\n--- Conclusion ---")
if p_val < ALPHA:
    print("Result: SIGNIFICANT DIFFERENCE")
    print(f"Since P-Value ({p_val:.5f}) < {ALPHA}, we REJECT the Null Hypothesis.")
    if acc_inc > acc_vgg:
        print(">> InceptionV3 significantly outperforms VGG16.")
    else:
        print(">> VGG16 significantly outperforms InceptionV3.")
else:
    print("Result: NO SIGNIFICANT DIFFERENCE")
    print(f"Since P-Value ({p_val:.5f}) > {ALPHA}, we FAIL TO REJECT the Null Hypothesis.")
    print(">> The models perform similarly on this test set.")
```
