```python
# ==============================================================================
# 0. SETUP & IMPORTS
# ==============================================================================
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from torchvision import transforms
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_curve, auc
from sklearn.preprocessing import label_binarize
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
import os
import random
import cv2
from PIL import Image
from tqdm.notebook import tqdm
from itertools import cycle
import subprocess

# Check/Install dependencies if missing
try:
    import torchinfo
    from torchinfo import summary
except ImportError:
    print("Installing torchinfo...")
    subprocess.check_call(['pip', 'install', 'torchinfo'])
    from torchinfo import summary

print("Libraries loaded.")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# ==============================================================================
# 1. CONFIGURATION
# ==============================================================================
# --- Variables ---
seed = 42
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

# --- Params ---
DATA_DIR = '/kaggle/input/project-dataset/Betel Leaf Image Dataset from Bangladesh/Original Images'
# Ensure the output dir is robust
OUTPUT_DIR = "/kaggle/working/processed-dataset-secnn"
os.makedirs(OUTPUT_DIR, exist_ok=True)

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
TEST_BATCH_SIZE = 16
EPOCHS = 50
PATIENCE = 10
LR = 0.001

# ==============================================================================
# 2. DATA LOADING & SPLITTING
# ==============================================================================
def load_and_preprocess(data_dir, output_dir):
    # Collect paths
    file_list = []
    classes = sorted(os.listdir(data_dir))
    
    for cls in tqdm(classes, desc="Scanning Dataset"):
        src_dir = os.path.join(data_dir, cls)
        if not os.path.isdir(src_dir): continue
            
        dest_dir = os.path.join(output_dir, cls)
        os.makedirs(dest_dir, exist_ok=True)
        
        for fname in os.listdir(src_dir):
            if not fname.lower().endswith(('.png', '.jpg', '.jpeg')): continue
            
            src_path = os.path.join(src_dir, fname)
            dst_path = os.path.join(dest_dir, fname)
            
            # Simple cache check
            if not os.path.exists(dst_path):
                try:
                    img = Image.open(src_path).convert('RGB')
                    img = img.resize(IMG_SIZE, Image.Resampling.LANCZOS)
                    img.save(dst_path, quality=95)
                    file_list.append((dst_path, cls))
                except Exception as e:
                    pass
            else:
                file_list.append((dst_path, cls))
    return pd.DataFrame(file_list, columns=['filepath', 'label']), classes

# Load & Process
df, class_names = load_and_preprocess(DATA_DIR, OUTPUT_DIR)
num_classes = len(class_names)
print(f"Total Images: {len(df)}")
print(f"Classes: {class_names}")

# Splits: 70% Train, 15% Val, 15% Test
train_df, temp_df = train_test_split(df, test_size=0.3, stratify=df['label'], random_state=seed)
val_df, test_df = train_test_split(temp_df, test_size=0.5, stratify=temp_df['label'], random_state=seed)

# Stats
n_train, n_val, n_test = len(train_df), len(val_df), len(test_df)
print(f"Train: {n_train} | Val: {n_val} | Test: {n_test}")

# --- Data Visualization (Matching Task 2 Style) ---
plt.figure(figsize=(15, 6))
# Pie Chart
plt.subplot(1, 2, 1)
plt.pie([n_train, n_test, n_val], labels=['Train', 'Test', 'Validation'], 
        colors=['#66b3ff', '#ff9999', '#99ff99'], autopct='%1.1f%%', startangle=140, shadow=True, explode=(0.05,0.05,0.05))
plt.title("Dataset Split Distribution")

# Bar Chart
plt.subplot(1, 2, 2)
counts = df['label'].value_counts().sort_index()
sns.barplot(x=counts.index, y=counts.values, palette='viridis')
plt.title('Class Distribution (Total)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# ==============================================================================
# 3. DATASET & DATALOADERS
# ==============================================================================
class BetelDataset(Dataset):
    def __init__(self, df, class_list, transform=None):
        self.df = df.reset_index(drop=True)
        self.transform = transform
        self.label_map = {cls: i for i, cls in enumerate(class_list)}
    
    def __len__(self): return len(self.df)
    
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = Image.open(row['filepath']).convert('RGB')
        if self.transform:
            img = self.transform(img)
        label = self.label_map[row['label']]
        return img, label

# Transforms
train_tf = transforms.Compose([
    transforms.RandomRotation(15),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.ColorJitter(0.2, 0.2),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])
test_tf = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

# Loaders
train_loader = DataLoader(BetelDataset(train_df, class_names, train_tf), batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
val_loader = DataLoader(BetelDataset(val_df, class_names, test_tf), batch_size=TEST_BATCH_SIZE, shuffle=False, num_workers=2)
test_loader = DataLoader(BetelDataset(test_df, class_names, test_tf), batch_size=TEST_BATCH_SIZE, shuffle=False, num_workers=2)


# ==============================================================================
# 4. SE-CustomCNN MODEL ARCHITECTURE (Task 4)
# ==============================================================================
print("\n--- Building SE-CustomCNN ---")

# --- Squeeze-and-Excitation Block ---
class SEBlock(nn.Module):
    def __init__(self, channel, reduction=16):
        super(SEBlock, self).__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channel, channel // reduction, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(channel // reduction, channel, bias=False),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _, _ = x.size()
        y = self.avg_pool(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1)
        return x * y.expand_as(x)

# --- Custom CNN with SE Blocks Integration ---
class SECustomCNN(nn.Module):
    def __init__(self, num_classes):
        super(SECustomCNN, self).__init__()
        
        # Block 1
        self.conv1 = nn.Conv2d(3, 32, 3, padding=1)
        self.bn1 = nn.BatchNorm2d(32)
        self.relu = nn.ReLU(inplace=True)
        self.se1 = SEBlock(32) # Integrate Attention
        self.pool1 = nn.MaxPool2d(2, 2)
        
        # Block 2
        self.conv2 = nn.Conv2d(32, 64, 3, padding=1)
        self.bn2 = nn.BatchNorm2d(64)
        self.se2 = SEBlock(64) # Integrate Attention
        self.pool2 = nn.MaxPool2d(2, 2)
        
        # Block 3
        self.conv3 = nn.Conv2d(64, 128, 3, padding=1)
        self.bn3 = nn.BatchNorm2d(128)
        self.se3 = SEBlock(128) # Integrate Attention
        self.pool3 = nn.MaxPool2d(2, 2)
        
        # Block 4 (Target for GradCAM)
        self.conv4 = nn.Conv2d(128, 256, 3, padding=1)
        self.bn4 = nn.BatchNorm2d(256)
        self.se4 = SEBlock(256) # Integrate Attention
        self.pool4 = nn.MaxPool2d(2, 2)
        
        # Classifier
        self.flatten = nn.Flatten()
        # Assuming 224 input -> 4 pools -> 14x14 spatial size
        self.fc1 = nn.Linear(256 * 14 * 14, 512) 
        self.dropout = nn.Dropout(0.5)
        self.fc2 = nn.Linear(512, num_classes)
        
    def forward(self, x):
        # Standard Conv flow with SE enhancement
        x = self.conv1(x)
        x = self.bn1(x)
        x = self.relu(x)
        x = self.se1(x)
        x = self.pool1(x)
        
        x = self.pool2(self.se2(self.relu(self.bn2(self.conv2(x)))))
        x = self.pool3(self.se3(self.relu(self.bn3(self.conv3(x)))))
        
        # Keep reference to last conv for CAM
        x = self.conv4(x)
        x = self.bn4(x)
        x = self.relu(x)
        x = self.se4(x) # Apply attention before final pool
        x = self.pool4(x)
        
        x = self.flatten(x)
        x = self.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        return x

model = SECustomCNN(num_classes).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=LR)
summary(model, input_size=(BATCH_SIZE, 3, 224, 224))


# ==============================================================================
# 5. TRAINING
# ==============================================================================
print(f"\n--- Starting Training ({EPOCHS} Epochs) ---")

history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
best_val_loss = float('inf')
patience_counter = 0

for epoch in range(EPOCHS):
    # Train
    model.train()
    t_loss, correct, total = 0, 0, 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS}", leave=False)
    
    for x, y in pbar:
        x, y = x.to(device), y.to(device)
        optimizer.zero_grad()
        out = model(x)
        loss = criterion(out, y)
        loss.backward()
        optimizer.step()
        
        t_loss += loss.item()
        _, p = torch.max(out, 1)
        total += y.size(0)
        correct += (p == y).sum().item()
        
    train_acc = 100 * correct / total
    train_loss = t_loss / len(train_loader)
    
    # Val
    model.eval()
    v_loss, correct, total = 0, 0, 0
    with torch.no_grad():
        for x, y in val_loader:
            x, y = x.to(device), y.to(device)
            out = model(x)
            v_loss += criterion(out, y).item()
            _, p = torch.max(out, 1)
            total += y.size(0)
            correct += (p == y).sum().item()
            
    val_acc = 100 * correct / total
    val_loss = v_loss / len(val_loader)
    
    # Log
    history['train_loss'].append(train_loss)
    history['train_acc'].append(train_acc)
    history['val_loss'].append(val_loss)
    history['val_acc'].append(val_acc)
    
    print(f"Ep {epoch+1}: Train Loss {train_loss:.4f} Acc {train_acc:.2f}% | Val Loss {val_loss:.4f} Acc {val_acc:.2f}%")
    
    # Early Stopping
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        patience_counter = 0
        torch.save(model.state_dict(), os.path.join(OUTPUT_DIR, 'se_cnn_best.pth'))
    else:
        patience_counter += 1
        if patience_counter >= PATIENCE:
            print("Early Stopping Triggered.")
            break

# Plot Training Curves
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history['train_acc'], label='Train')
plt.plot(history['val_acc'], label='Val')
plt.title("Accuracy")
plt.legend()
plt.subplot(1, 2, 2)
plt.plot(history['train_loss'], label='Train')
plt.plot(history['val_loss'], label='Val')
plt.title("Loss")
plt.legend()
plt.show()


# ==============================================================================
# 6. EVALUATION (TEST SET)
# ==============================================================================
print("\n--- Final Evaluation ---")
model.load_state_dict(torch.load(os.path.join(OUTPUT_DIR, 'se_cnn_best.pth')))
model.eval()

y_true, y_score, y_pred = [], [], []

with torch.no_grad():
    for x, y in tqdm(test_loader, desc="Testing"):
        x = x.to(device)
        out = model(x)
        probs = F.softmax(out, dim=1)
        _, p = torch.max(out, 1)
        
        y_true.extend(y.numpy())
        y_score.extend(probs.cpu().numpy())
        y_pred.extend(p.cpu().numpy())

# Metrics
print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title("CM (Counts)")
plt.subplot(1, 2, 2)
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title("CM (Normalized)")
plt.tight_layout(); plt.show()

# ROC / AUC
print("Generating ROC Curves...")
y_true_bin = label_binarize(y_true, classes=range(num_classes))
y_score = np.array(y_score)

plt.figure(figsize=(10, 8))
colors = cycle(['blue', 'red', 'green', 'orange'])
for i, color in zip(range(num_classes), colors):
    fpr, tpr, _ = roc_curve(y_true_bin[:, i], y_score[:, i])
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, color=color, lw=2, label=f'{class_names[i]} (AUC = {roc_auc:.3f})')

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
plt.title('ROC Curves: SE-CustomCNN')
plt.legend(loc="lower right"); plt.grid(alpha=0.3)
plt.show()


# ==============================================================================
# 7. GRAD-CAM EXPLAINABILITY
# ==============================================================================
class GradCAM:
    def __init__(self, model, layer):
        self.model = model
        self.layer = layer
        self.grads = None
        self.acts = None
        # Hooks (using basic forward/backward hooks for simplicity & compat)
        self.handles = [
            self.layer.register_forward_hook(self.f_hook),
            self.layer.register_full_backward_hook(self.b_hook)
        ]
    def f_hook(self, m, i, o): self.acts = o
    def b_hook(self, m, gi, go): self.grads = go[0]
    
    def __call__(self, x):
        self.model.zero_grad()
        out = self.model(x)
        idx = out.argmax(1).item()
        out[:, idx].backward()
        
        pooled_grads = torch.mean(self.grads, dim=[0, 2, 3])
        cam = (self.acts * pooled_grads.view(1, -1, 1, 1)).sum(1)
        cam = F.relu(cam)
        cam = cam - cam.min(); cam = cam / (cam.max() + 1e-8)
        return cam.detach().cpu().numpy()[0], idx

    def remove(self):
        for h in self.handles: h.remove()

print("\n--- Visualizing Attention (Grad-CAM) ---")
# Denormalize func
denorm = lambda x: x.cpu().permute(1,2,0) * torch.tensor([0.229,0.224,0.225]) + torch.tensor([0.485,0.456,0.406])

# Init CAM on last conv layer (self.conv4 in our SECustomCNN)
grad_cam = GradCAM(model, model.conv4)

# Get Samples (One from each class ideally, but random batch here)
iter_test = iter(test_loader)
imgs, lbls = next(iter_test)

plt.figure(figsize=(12, 12))
# Show first 4 images
for i in range(min(4, len(imgs))):
    x_in = imgs[i:i+1].to(device)
    true_cls = class_names[lbls[i]]
    
    mask, pred_idx = grad_cam(x_in)
    pred_cls = class_names[pred_idx]
    
    # Prepare images
    orig = np.clip(denorm(imgs[i]).numpy(), 0, 1)
    
    mask_rez = cv2.resize(mask, (224, 224))
    heatmap = cv2.applyColorMap(np.uint8(255 * mask_rez), cv2.COLORMAP_JET)
    heatmap = cv2.cvtColor(heatmap, cv2.COLOR_BGR2RGB)
    
    overlay = cv2.addWeighted(np.uint8(255*orig), 0.6, heatmap, 0.4, 0)

    # Plot row
    plt.subplot(4, 3, i*3 + 1)
    plt.imshow(orig); plt.axis('off'); plt.title(f"True: {true_cls}")
    plt.subplot(4, 3, i*3 + 2)
    plt.imshow(heatmap); plt.axis('off'); plt.title("SE-Attention Heatmap")
    plt.subplot(4, 3, i*3 + 3)
    plt.imshow(overlay); plt.axis('off'); plt.title(f"Pred: {pred_cls}", color='g' if true_cls==pred_cls else 'r')

plt.tight_layout()
plt.show()

grad_cam.remove()
```
